<summary>Cliquer pour voir la solution</summary>

<pre><code>def somme_for(n):
    somme = 0
    for i in range(n+1):
        somme += 2**i
    return somme

def somme_while(n):
    somme = 1
    i = 0
    while i < n:
        i += 1
        somme += 2**i
    return somme

def somme_rec(n):
    if n == 0:
        return 1
    else:
        return somme_rec(n-1) + 2**n
