## Triangle de Sierpiński

```python
import matplotlib.pyplot as plt
from random import randint, random
from math import sqrt

listes_sommets = [0, 1, complex(1/2, sqrt(3)/2)]



def sierpinski(n):
    coefficients = [random() for _ in range(3)]
    point = sum(c * s for c, s in zip(coefficients, listes_sommets))/sum(coefficients)
    liste_points = [point]
    for _ in range(n):
        point = (point + listes_sommets[randint(0, 2)]) / 2
        liste_points.append(point)
    plt.scatter([p.real for p in liste_points], [p.imag for p in liste_points], color='red', s=1)
    plt.axis("equal")
    plt.show()

sierpinski(10000)
```

```python
import matplotlib.pyplot as plt

def nouveau_triangle(triangle):
    A, B, C = triangle
    AB = (A + B) / 2
    AC = (A + C) / 2
    BC = (B + C) / 2
    return [A, AB, AC], [AB, B, BC], [AC, BC, C]

def nouvelle_liste_triangles(liste_triangles):
    new_triangles = []
    for t in liste_triangles:
        new_triangles.extend(nouveau_triangle(t))
    return new_triangles

def sierpinski_rec( n):
    triangle = [0, 1, complex(1/2, sqrt(3)/2)]
    liste_triangles = [triangle]
    for _ in range(n):
        liste_triangles = nouvelle_liste_triangles(liste_triangles)
    plt.figure(figsize=(6, 6))
    for t in liste_triangles:
        plt.fill([p.real for p in t], [p.imag for p in t], color='red')
    plt.axis("equal")
    plt.show()

sierpinski_rec(6)
```
