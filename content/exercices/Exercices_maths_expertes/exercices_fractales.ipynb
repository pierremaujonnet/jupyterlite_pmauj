## Triangle de Sierpiński

```python
import matplotlib.pyplot as plt
from random import randint, random
from math import sqrt

listes_sommets = [0, 1, complex(1/2, sqrt(3)/2)]

coefficients = [random() for _ in range(3)]
sommet = sum(c * s for c, s in zip(coefficients, listes_sommets))/sum(coefficients)

def plot_triangle(n):
    for _ in range(n):
        plt.scatter(sommet.real, sommet.imag, color='blue', s=1)
        sommet = (sommet + listes_sommets[randint(0, 2)]) / 2
    plt.show()

plot_triangle(10)
```
