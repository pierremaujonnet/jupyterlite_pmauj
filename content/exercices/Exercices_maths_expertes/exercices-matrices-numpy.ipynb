**Exercice :** On considère les matrices $A=\left(\begin{array}{ccccc}5&3&1\\1&0&-3\\7&-5&1\\\end{array}\right)$ et $B=\left(\begin{array}{ccccc}0&2&-3\\2&1&-2\\0&4&-3\\\end{array}\right)$. 
1. Entrer $A$ et $B$ sous forme de tableaux numpy.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>import numpy as np
A = np.array([[5,3,1],[1,0,-3],[7,-5,1]])
B = np.array([[0,2,-3],[2,1,-2],[0,4,-3]])
</code></pre>
</details>

2. Calculer les matrices $A+5B$ et $A^t-3Id_3$.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>print(A+5*B)
print(A.T-3*np.eye(3))
</code></pre>
</details>

3. Calculer $A\times B$.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>print(A.dot(B))
</code></pre>
</details>

4. Calculer $A^5$.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>import numpy.linalg as alg
print(alg.matrix_power(A, 5))
</code></pre>
</details>

5. Calculer le déterminant de $A$. $A$ est-elle inversible ? Si oui, calculer $A^{-1}$.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>print(alg.det(A))
print(alg.inv(A))
</code></pre>
</details>

5. Les suites $(u_n)$ et $(v_n)$ sont définies par $\left\{\begin{array}{l}u_0=1\\v_0=1\\u_{n+1}=2u_n+3v_n\\v_{n+1}=u_n+v_n\\\end{array}\right.$.
		Calculer $u_1$, $v_1$, $u_2$, $v_2$, $u_{20}$ et $v_{20}$.

<details>
<summary>Cliquer pour voir la solution</summary>
<pre><code>U = np.array([[1],[1]])
A = np.array([[2,3],[1,1]])
print(A.dot(U))
print(alg.matrix_power(A,2).dot(U))
print(alg.matrix_power(A,20).dot(U))
</code></pre>
</details>

Sous `numpy`, en cas d'entiers trop grands, un calcul tel que `print(alg.matrix_power(A,100).dot(U))` sera fait modulo $2^{64}$ et entraînera de fait des erreurs.

On peut remédier à ce problème en utilisant des matrices sous la forme `U = np.array([[1], [1]], dtype=object)` et `A = np.array([[2, 3], [1, 1]], dtype=object)`.
