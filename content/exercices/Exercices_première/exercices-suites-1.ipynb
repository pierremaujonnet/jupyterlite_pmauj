**Exercice :** On considère la suite $(u_n)$ définie par $\left\{\begin{array}{l}
	u_0=1\\u_{n+1}=3u_n-1\\
\end{array}\right.$.

1. Décrire les rôles respectifs des fonctions suivantes :

```python
def u1(n):
    i=0
    u=1
    while i<n:
        u=3*u-1
        i=i+1
    return u

def seuil(M):
    n=0
    u=1
    while u<=M:
        u=3*u-1
        n=n+1
    return n

def Somme(n):
    i=0
    u=1
    S=u
    while i<n:
        u=3*u-1
        S=S+u
        i=i+1
    return S
```

**Remarque importante :** Il est également possible de renvoyer la valeur de $u(n)$ en fonction de $n$ à l'aide d'une fonction **récursive**.

```python
def u_rec(n):
    if n == 0:
        return 1
    else:
        return 3*u_rec(n-1)-1

print(u_rec(3))    
```
