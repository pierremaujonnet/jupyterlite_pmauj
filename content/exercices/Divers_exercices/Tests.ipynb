```python
import ipyturtle3 as turtle
from IPython.display import display

canvas = turtle.Canvas(width=500, height=350)
display(canvas)

ecran = turtle.TurtleScreen(canvas)
t = turtle.Turtle(ecran)
t.pencolor("blue")

for cote in range(4):
    t.forward(100)
    t.left(90)
```
