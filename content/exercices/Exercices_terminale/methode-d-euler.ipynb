# Méthode d'Euler
### Equation différentielle $y'=y$
On considère la solution $f$ de l'équation différentielle $y'=y$ telle que $f(0)=1$.

```python
import matplotlib.pyplot as plt
from math import exp

def approx_exp(x_max, n):
    h = x_max / n
    x = 0
    y = 1
    x_list = [x]
    y_list = [y]
    for _ in range(n):
        x = x + h
        y = y * (1+h)
        x_list.append(x)
        y_list.append(y)
    plt.plot(x_list,y_list)

def trace_exp(x_max, npoints=100):
    x_liste = [i/npoints for i in range(int(x_max*npoints)+1)]
    y_liste = [exp(x) for x in x_liste]
    plt.plot(x_liste, y_liste, 'r')

trace_exp(3)
approx_exp(3, 10)
approx_exp(3, 20)
approx_exp(3, 100)
plt.show()
```

### Equation différentielle $y'=-2y+4$
Soit $g$ la solution sur $\mathbb{R}$ de l'équation différentielle $y'=-2y+4$ telle que $g(0)=3$.

```python
import matplotlib.pyplot as plt
from math import exp

def methode_Euler(x_max, n):
    h = x_max / n
    x = 0
    y = 3
    x_liste = [x]
    y_liste = [y]
    for i in range(n):
        yp = -2*y+4
        x = x + h
        y = y + h*yp
        x_liste.append(x)
        y_liste.append(y)
    plt.plot(x_liste, y_liste)
    
def courbe_exacte(x_max, npoints=100):
    x_liste = [i/npoints for i in range(int(x_max*npoints)+1)]
    y_liste = [exp(-2*x)+2 for x in x_liste]
    plt.plot(x_liste, y_liste, 'r')

courbe_exacte(5)
methode_Euler(5, 10)
methode_Euler(5, 20)
methode_Euler(5, 100)

plt.show()
```
