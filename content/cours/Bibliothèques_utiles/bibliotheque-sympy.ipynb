# Qu'est-ce que SymPy ?

**SymPy *(Symbolic Python)*** est une bibliothèque Python destinée au **calcul formel**.

On l'importe usuellement via la commande : `import sympy as sp`

# Calculs symboliques de base

## Symboles/variables symboliques

Pour utiliser les fonctionnalités de calcul formel, il faut tout d’abord définir des variables symboliques. On peut
procéder de deux manières différentes.

```python
import sympy as sp

# Méthode 1
x, y = sp.symbols(' x y ' )

# Méthode 2
from sympy.abc import x, y
```

## Expressions symboliques

Les expressions symboliques sont créées à partir de variables symboliques. On 

```python
expr = x**2 + 2*x + 1
```

## Constantes/fonctions mathématiques

- `sp.pi `
- `sp.E`   
- `sp.I`      
- `sp.oo`  
- `-sp.oo`
- `sp.sqrt`
- `sp.cos`
- ...

**Remarque :** Les calculs sont par défaut exacts. Pour des valeurs approchées, utiliser les fonctions `evalf()` ou `N`.

```python
a = sp.sqrt(2)
print(a.evalf())
print(sp.N(a, 20))
```

##  Substitution et simplification d’expressions

- `expr.subs(x, valeur)` 
- `sp.simplify(expr)`
- `sp.factor(expr) `: factorisation
- `sp.expand(expr)` : développement

## Fractions

Pour calculer avec des nombres rationnels exacts, on peut utiliser `sp.S` ou `sp.Rational`.

```python
print(sp.S(2)/3)
print(sp.Rational(2,3))
print(4+sp.Rational(2,3))
```

On peut également simplifier des expressions littérales rationnelles

```python
f = (x**2 - 4) / (x**2 - 3*x + 2)
g = 1/(x+1) - 3/(x-1)
h = (2*x+3)/(x**3-2*x**2+x)

print(f"Simplification : {sp.cancel(f)}")
print(f"Mise sous dénominateur commun : {sp.together(g)}")
print(f"Décomposition en éléments simples : {sp.apart(h)}")
```

## Dérivation/intégration/limites

```python
print(sp.diff(expr, x)) # dérivée
print(sp.diff(sp.sin(x), x)) # dérivée
print(sp.diff(expr, x,2)) # dérivée seconde
print(sp.integrate(2*x + 3, x)) # primitive
print(sp.integrate(2*x, (x, 0, 2))) # intégrale
print(sp.limit(sp.sin(x)/x, x, 0)) # limite
```

## Equations/équations différentielles

```python
print(sp.solve(sp.Eq(x**2 + x - 2, 0), x))

print(sp.solveset( x**2 - 2, x, domain=sp.S.Reals))

equation = x**2 - 4
solution = sp.solve(equation, x)
print(solution)

eq1 = x + y - 2
eq2 = x - y 
solution_systeme = sp.solve((eq1, eq2), (x, y))
print(solution_systeme)

Y = sp.Function('Y' )
diff_eq = sp.diff(Y(x), x, 2) + Y(x)
solution_diff = sp.dsolve(diff_eq)
print(solution_diff)
```

# Matrices

```python
from sympy import Matrix, det, zeros, eye, ones
from sympy.abc import a, b, c, d

# Création de matrices
M = Matrix([[1, 2], [3, 4]])
# Matrice symbolique
A = Matrix([[a, b], [c, d]])
```

```python
# Transposée
print(M.transpose())

# Déterminant
print(M.det())

# Trace
print(M.trace())

# Inverse
print(M.inv())

# Addition de matrices
N = Matrix([[5, 6], [7, 8]])
print(M + N)

# Multiplication matricielle
print(M * N)

# Multiplication par un scalaire
print(2 * M)

# Résolution de systèmes linéaires
b = Matrix([1, 2])

solution = M.solve(b)
print(f"Solution de Mx = b : {solution}" )
```

```python
# Matrice nulle 2x2
zero_2x2 = zeros(2)
print(f" Matrice nulle 2x2: \n{zero_2x2}" )

# Matrice nulle 2x3
zero_2x3 = zeros(2, 3)

# Autre méthode avec Matrix
zero_alt = Matrix.zeros(2, 3)

# Matrice identité 3x3
id_3x3 = eye(3)

# Autre méthode avec Matrix
id_alt = Matrix.eye(3)

# Tous les éléments sont 1
ones_matrix = ones(2, 3)

# Création d'une matrice diagonale à partir d'une liste
diagonal_values = [1, 2, 3]
diagonal_matrix = sp.diag(*diagonal_values)
```
