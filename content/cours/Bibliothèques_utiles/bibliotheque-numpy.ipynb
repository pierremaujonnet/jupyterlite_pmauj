# Qu'est-ce que NumPy ?

**NumPy *(Numerical Python)*** est une bibliothèque Python destinée au **calcul en sciences**.

On l'importe usuellement via la commande : `import numpy as np`

# Tableaux Numpy

## Tableaux et `dtype`

Les **tableaux** numpy peuvent être à une dimension (listes), à deux dimensions (matrices) ou même plus ...

- Exemple de tableau de dimension 1 : `np.array([3.2, 5.01, -8, 9.12, -3.56, 4])`
- Exemple de tableau de dimension 2 : `np.array([[2,5,8,9],[3,4,1,0]])`
- Exemple de tableau de dimension 3 : `np.array([[[2,5,8,9],[3,4,1,0]],[[-1,2,3,5],[0,1,0,2]]])`

Pour optimiser la mémoire et augmenter la vitesse de calcul, tous les éléments d’un tableau numpy doivent contenir des éléments de même type, le `dtype` qui indique l'espace mémoire nécessaire à chaque élément.

```python
import numpy as np

a = np.array([3.2, 5.01, -8, 9.12, -3.56, 4])
b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])
c = np.array([[[2,5,8,9],[3,4,1,0]],[[-1,2,3,5],[0,1,0,2]]])
d = np.array([[True, False, True, True],[False, False, True, True],[True, False, False, False]])

print("a : ", a)
print("b : ", b)
print("c : ", c)
print("d : ", d)

print("Type de a : ", type(a))
print("dtype des éléments de a : ", a.dtype)
print("dtype des éléments de b : ", b.dtype)
print("dtype des éléments de c : ", c.dtype)
print("dtype des éléments de d : ", d.dtype)
```

## Dimensions, espaces mémoire

Soit `A` un tableau numpy.
- `A.ndim` : nombre de dimensions 
- `A.shape` : dimensions sous forme de *tuple*
- `A.size` : nombre total d'éléments
- `A.itemsize` : espace mémoire d'un élément, en octets *(en relation avec `A.dtype`)*
- `A.nbytes` : espace mémoire total des données, en octets

```python
import numpy as np

a = np.array([3.2, 5.01, -8, 9.12, -3.56, 4])
b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])
c = np.array([[[2,5,8,9],[3,4,1,0]],[[-1,2,3,5],[0,1,0,2]]])
d = np.array([[True, False, True, True],[False, False, True, True],[True, False, False, False]])

print("Nombre de dimensions de a : ", a.ndim)
print("Nombre de dimensions de b : ", b.ndim)
print("Nombre de dimensions de c : ", c.ndim)
print("Nombre de dimensions de d : ", d.ndim)
print("Dimensions de a : ", a.shape)
print("Dimensions de b : ", b.shape)
print("Dimensions de c : ", c.shape)
print("Dimensions de d : ", d.shape)
print("Nombre de lignes de b : ", b.shape[0])
print("Nombre de colonnes de b : ", b.shape[1])
print("Nombre total d'éléments de b : ", b.size)
print("Espace mémoire total des données de b : ", b.nbytes,"o")
print("Espace mémoire d'un élément de a : ", a.itemsize,"o")
print("Espace mémoire d'un élément de b : ", b.itemsize,"o")
print("Espace mémoire d'un élément de d : ", d.itemsize,"o")
```

## Tableaux particuliers

- `np.zeros(format)` : tableau composé uniquement de 0
- `np.ones(format)` : tableau composé uniquement de 1
- `np.full(format, element)` : tableau composé de la répétition d'un élément
- `np.eye(n)` : matrice unité d'ordre $n$
- `np.indices(format)` : grilles des indices

```python
import numpy as np

print(np.zeros(5))
print(np.zeros((3, 6)))
print(np.ones((3, 6)))
print(np.full((2, 4), True))
print(np.eye(4))
```

```python
print(np.indices((3,4)))
```

- `np.arange(debut, fin, pas)` : tableau à 1 dimension de valeurs régulièrement espacées d'un pas fixé
- `np.linspace(debut, fin, nombre)` : tableau à 1 dimension d'un nombre fixé de valeurs régulièrement espacées

```python
import numpy as np

print(np.arange(-1, 4, 0.5))
print(np.linspace(-6, 6, 11))
```

## Eléments et slicing

- *1 dimension* : similaire aux listes en Python
    - A[i]
    - A[debut:fin:pas]
    - A[:fin]
    - A[debut:]
    - ...
- *2 dimensions* :
    - A[i,j] : élément à l’intersection de la i + 1-ième ligne et de la j + 1-ième colonne
        
        Les lignes et les colonnes sont numérotées à partir de l'indice 0.
        
    - A[i,:]/A[:,j] : i + 1-ième ligne/j + 1-ième colonne
    - A[i,:].copy()/A[:,j].copy() : copie de la i + 1-ième ligne/j + 1-ième colonne
    - ...
- ...

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

print(b[0,0])
print(b[1,2])
print("Première ligne : ",b[0,:])
print("Troisième colonne : ",b[:,2])
```

# Opérations

## Calculs &laquo; vectorisés &raquo;

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])
e = np.array([[6, 7, 10, 9.1],[4, 2, 5.5, 3.2]])

print(b+e)
print(5*b)
print(3*b-5*e)
print(b**2)
print(e**5)
```

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])
e = np.array([[6, 7, 10, 9.1],[4, 2, 5.5, 3.2]])

print(np.sqrt(e))
print(np.cos(b))
print(np.sin(b))
print(np.exp(b))
print(np.log(e))
```

**Remarque importante :** Les constantes mathématiques $\pi$ et e s'obtiennent par `np.pi` et `np.e`.

##  Maximum, minimum, somme, moyenne, variance, écart-type

### Sur le tableau entier

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

print(b.max())
print(b.argmax())
print(b.min())
print(b.argmin())
print(b.sum())
print(b.mean())
print(b.var())
print(b.std())
```

### Selon une direction

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

print(b.max(axis=0))
print(b.max(axis=1))
print(b.sum(axis=0))
print(b.sum(axis=1))
print(b.mean(axis=0))
print(b.mean(axis=1))
```

## Booléens

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

print(b>2)
print(~(b>2))
print((b<=4) & (b>-2))
print((b>2) | (b<-3))
print(b[b>2])
```

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

b[b>2] = 0
print(b)
```

```python
import numpy as np

b = np.array([[2, -5, 8, 9],[-3, 4, 1, 0]])

print(np.where(b>2)) # sous forme de tuple
print(np.argwhere(b>2)) # sous forme de tableau
```

# Un peu d'aléatoire

## Concaténation, répétition, nombre de caractères

Il est possible de produire des tableaux pseudo-aléatoires via `random`.

```python
import numpy as np

print(np.random.rand()) # flottant entre 0 et 1
print(np.random.rand(6)) 
print(np.random.rand(4,3))
print(np.random.randint(1, 10)) # entiers 10 est exclu
print(np.random.randint(1, 10, 8)) # entiers 10 est exclu
```

```python
import numpy as np

tab = np.array([1, 2, 3, 4, 5, 6])

# Choix au hasard dans un tableau unidimensionnel
print(np.random.choice(tab))
print(np.random.choice(tab,10))

# Mélange au hasard des éléments en place
np.random.shuffle(tab)         
print(tab)
```

**Remarques importantes :** 
- Il est possible de fixer une **graine** par `np.random.seed(graine)`.
- De manière plus moderne, on posera `rng = np.random.default_rng(n)`. Les fonctions sont alors :
    - `rng.random(n)`
    - `rng.integers(debut, fin, nbre)` 
    - ...

# Calcul matriciel

## Produit matriciel

Sous condition d'existence, le produit matriciel de deux matrices $A$ et $B$ modélisées sous numpy se fait par les instructions `np.dot(A, B)` ou `A.dot(B)`.

```python
import numpy as np

A = np.array([[2,1,4],[5,3,2],[-2,3,0]])
B = np.array([[3,1,6],[-1,0,4],[7,4,11]])

print(A.dot(B))
```

## Transposée

La transposée d'une matrice $A$ se fait via l'instruction `A.T`.

## Module d'algèbre linéaire

Peut être importé via l'instruction `import numpy.linalg as alg`.
- `alg.matrix_power(A,n)` : matrice An
- `alg.det(A)` : déterminant de la matrice A
- `alg.inv(A)` : inverse de la matrice A (sous condition d’existence)
- `alg.solve(A,B)` : solution de l’équation AX = B (sous condition d’existence)

```python
import numpy as np
import numpy.linalg as alg

A = np.array([[2,1,4],[5,3,2],[-2,3,0]])
C = np.array([[-1],[4],[7]])

print(alg.matrix_power(A,5))
print(alg.det(A))
print(alg.inv(A))
print(alg.solve(A,C))
```
