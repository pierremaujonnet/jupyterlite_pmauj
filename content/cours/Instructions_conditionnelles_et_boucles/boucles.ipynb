# Boucle `while`

### Principe

**Tant qu'une condition est réalisée**, on éxécute une suite d'instructions, ce qui déclenche une *boucle*.

On ne sort de la boucle que lorsque la condition n'est plus réalisée.

### Condition d'arrêt

Lorsque lorsque l'on veut déterminée *une valeur qui vérifie une condition*, on doit indiquée la **négation de cette condition** au niveau de `while`.

**Exemple :** Déterminer le plus petit entier naturel $n$ tel que $\left(n+1\right)\left(\sqrt{n}-2\right)>10000$.

````python
from math import sqrt

n = 0
while (n+1)*(sqrt(n)-2) <= 10000:
    # La négation de (n+1)*(sqrt(n)-2)>10000 est (n+1)*(sqrt(n)-2) <= 10000
    n = n+1
    
print(n)
````

### Boucle infinie

Si la condition d'arrêt est *toujours réalisée*, on ne sort pas de la boucle qui est **éxécutée infiniment**.

On évite en général les boucles infinies qui vont nécessité de forcer l'arrêt de l'éxécution de votre programme, l'arrêt du noyau voire l'arrêt du logiciel.

⚠️ Exemple de boucle infinie à ne pas exécuter :

```
n = 0
while n >= 0:
    n = n+1
```


Il existe des exceptions à ce qui précède : lorsque *la sortie de la boucle peut être déclenchée par l'utilisateur*.

````python
n = 0

while True :
    n = n + 1
    print("n = ",n)
    reponse = input("Continuer ? o/n\n")
    
    if reponse == "n":
        break
    
print("La boucle est terminée !")    
````

# Boucle`for`

### Mot-clé `for`

Le mot-clé `for` permet de parcourir des objets **itérables** tels des *chaînes de caractères* ou des *listes*.

````python
mot = "Bonjour"

for lettre in mot:
    print(lettre)
````

````python
liste = [1, -3, 4, 10, -2]

for x in liste:
    print(x**2)
````

### Mot-clé `range`

Le mot-clé `range` crée un objet itérable énumérant des entiers relatifs situés à des écarts réguliers.

La syntaxe générale est `range(debut, fin, pas)`. 
- `debut`, `fin` et `pas` sont des entiers relatifs **positifs ou négatifs**.
- `debut` est **inclus** et `fin` est **exclu**.
- L'ensemble des entiers considérés peut être vide selon les valeurs (en particulier selon le signe de `pas`).

*Pour observer les résultats produits, nous allons convertir les objets itérables en listes, ce qui n'est pas nécessaire dans de nombreux cas d'utilisation.*

````python
print(list(range(2, 40, 3)))
print(list(range(2, 24, 2)))
print(list(range(-50, -2, 5)))
print(list(range(30, -11, -7)))
````

````python
print(list(range(11, 3, 2)))
print(list(range(2, 18, -5)))
````

**Cas particuliers :**
- Si le pas est de 1, on note plus simplement : `range(debut, fin)`.
- Si le pas est de 1 et si la première valeur est 0, on note plus simplement : `range(fin)`.

````python
print(list(range(2, 13)))
print(list(range(8)))
````

````python
for i in range(1, 11):
    print(i**2)
````

### Imbrications de boucles

````python
for i in range(1, 11):
    for j in range(1, 11):
        print(i," * ", j, " = ", i*j)
````

# Exemples importants : calculs de sommes

### Calculer une somme à l'aide de `for`

Nous souhaitons calculer $5+7+9+11+\cdots+1023$.

````python
somme = 0
for i in range(5, 1024, 2):
    somme = somme + i

print(somme)
````

### Calculer une somme à l'aide de `while`

````python
somme = 0
i = 5
while i <= 1023:
    somme = somme + i
    i = i + 2

print(somme)
````

### Exemple classique

Ecrire une fonction renvoyant $1^2+2^2+3^2+4^2+\cdots+n^2$ où $n$ est un entier naturel non nul.

````python
# Utilisation de for
def somme1(n):
    somme = 0 # initialisation de la valeur de la somme à 0
    for i in range(1, n+1): # La dernière valeur n'est pas comprise dans l'itération donc on remplace n par n+1
        somme = somme + i**2
    return somme

# Utilisation de while
def somme2(n):
    somme = 0
    i = 1
    while i <= n: # Bien penser au inférieur ou égal
        somme = somme + i**2
        i = i + 1 # Incrémenter i à la fin  de la boucle
    return somme

print(somme1(10))
print(somme2(10))
````
