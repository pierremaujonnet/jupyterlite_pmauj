# Instructions conditionnelles et variables booléennes

## Instructions conditionnelles

En informatique, une **structure conditionnelle** effectue une suite d'instructions si une condition (ou un ensemble de conditions) définie est remplie (on parle de *variable booléenne vraie*) et, éventuellement, une autre suite d'instructions si elle n'est pas remplie (on parle de *variable booléenne fausse*). Le schéma est ainsi :
$$\begin{array}{rcl}\text{Si condition}&\to&\text{suite d'instructions 1}\\\text{Sinon}&\to&\text{suite d'instructions 2}\end{array}$$

En langage Python, on utilise les mots-clés `if` et `else`.

```python
def est_divisible_par_3(n):
    """
    Fonction permettant de déterminer si un entier relatif est divisible par 3.
    """
    if n%3 == 0:
        return "n est divisible par 3."
    else:
        return "n n'est pas divisible par 3."
```

**Remarque importante :** On a utilisé le fait qu'un entier relatif $n$ est divisible par 3 si et seulement si son reste dans la division euclidienne par 3 est 0.

Plus généralement, en langage Python, si $a$ et $b$ sont deux entiers relatifs avec $a\neq0$, la condition &laquo; $a$ divise $b$ &raquo; s'écrira : `b%a == 0`.

## Conditions multiples

Lorsque l'on a besoin de faire intervenir des **conditions alternatives**, on utilise le mot-clé `elif` *(contraction de else et de if)*.

```python
def est_rectangle(a, b, c):
    """
    Fonction permettant de déterminer si un triangle ABC est rectangle,
    les longueurs a=BC, b=AC et c=AB étant données.
    """
    if a**2 == b**2 + c**2:
        return "Le triangle ABC est rectangle en A."
    elif b**2 == a**2 + c**2:
        return "Le triangle ABC est rectangle en B."
    elif c**2 == a**2 + b**2:
        return "Le triangle ABC est rectangle en C."
    else:
        return "Le triangle ABC n'est pas rectangle."
```

## Variables booléennes

### Nature d'une variable booléenne

Une **variable booléenne** peut prendre deux valeurs : Vrai ou Faux. Ainsi &laquo; $2>3$ &raquo; prend la valeur Faux.

En langage Python : `True` et `False` (Attention à la majuscule !)

```python
2 > 3
```

```python
5 <= 5
```

```python
2 > 3 and 5 <= 5
```

```python
2 > 3 or 5 <= 5
```

### Opérations entre variables booléennes

Soient P et Q deux variables booléennes.
    <ul>
        <li><code>P and Q</code> : P <strong>et</strong> Q</li>
        <li><code>P or Q</code> : P <strong>ou</strong> Q</li>
        <li><code>not P</code> : <strong>négation de</strong> P</li>
    </ul></div>

```python
def est_equilateral(a, b, c):
    """
    Fonction permettant de déterminer si un triangle ABC est équilatéral,
    les longueurs a=BC, b=AC et c=AB étant données.
    """
    if a == b and a == c:
        return "Le triangle ABC est équilatéral."
    else:
        return "Le triangle ABC n'est pas équilatéral."
```
