# Initiation à Matplotlib

## Qu'est-ce que Matplotlib ?

Matplotlib (Mathematical Plotting Library) est une bibliothèque Python destinée aux représentations graphiques statiques ou animées.

On l'importe usuellement via la commande : `import matplotlib.pyplot as plt`

## Tracés de base

### Nuages de points

Un nuage de points est représenté à l'aide de la fonction `scatter` et de deux listes (listes Python ou tableaux numpy de dimension 1) :
- la **liste des abscisses** des points du nuage ;
- la **liste des ordonnées** des points du nuage.

Ainsi, les listes doivent contenir le **même nombre d'éléments**. De plus, l'ordre des coordonnées dans la liste, sera l'ordre de tracé.

```python
import matplotlib.pyplot as plt

x = [1, 2,  3, 4, 5]
y = [-2, 1, -5, 3, 0]

plt.scatter(x, y)

plt.show() # affiche la figure
```

### Courbes

Une courbe sous matplotlib est un ensemble de points reliés par des segments de droite. Elle est représentée à l'aide de la fonction `plot` et de deux listes (listes Python ou tableaux numpy de dimension 1), de manière similaire aux nuages de points.

```python
import matplotlib.pyplot as plt
x = [1, 2,  10, 4, 5]
y = [-2, 1, -5, 3, 0]

plt.plot(x, y)

plt.show() # affiche la figure
```

**Cas de la représentation graphique d'une fonction :**

```python
import matplotlib.pyplot as plt

def f(t):
    return 2*t**2-5*t+1

x = list(range(-5, 6))
y = [f(t) for t in x]

plt.plot(x, y)

plt.show() # affiche la figure
```

### Utilisation de NumPy

Pour une courbe représentative de fonction, l'utilisation de NumPy peut être particulièrement adaptée.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1)
plt.plot(x, y2)

plt.show() # affiche la figure
```

### Personnalisation

#### Style et couleurs

En ajoutant des arguments dans `plt.scatter` ou `plt.plot`, on peut adapter couleurs, marqueur, style du tracé, épaisseur du tracé, ...

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)
y3 = np.cos(x) + np.sin(x)

plt.plot(x, y1, color="red", linewidth=3, linestyle="--")
plt.plot(x, y2, color="#DB56FF", marker = "o")
plt.plot(x, y3, color="teal", linewidth=3, linestyle=":")

plt.show() # affiche la figure
```

#### Ajouter une légende

Pour associer des légendes aux courbes, on ajoute des valeurs aux attributs `label` et on utilise la fonction `legend` avec comme argument optionnel la position.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1, label="Fonction cosinus")
plt.plot(x, y2, label="Fonction sinus")

# Légende placée en haut à gauche
plt.legend(loc="upper left")

plt.show() # affiche la figure
```

#### Axes

Pour personnaliser les axes, on peut les récupérer via la commande `plt.gca()`.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1)
plt.plot(x, y2)

# Récupération des axes
ax = plt.gca()

# Limites des axes
ax.set_xlim(-6.5, 6.5) 
ax.set_ylim(-1.8, 2)

# Titrage des axes
ax.set_title("Fonctions trigonométriques") 
ax.set_xlabel("Abscisses") 
ax.set_ylabel("Ordonnées")

# Ajout d'une grille personnalisable
ax.grid( visible=True, linestyle=":", linewidth=0.8, alpha=0.7 )

# Graduations sous forme de listes ou de tableaux numpy de dimension 1
ax.set_xticks(np.arange(-6, 7, 2)) 
ax.set_yticks([-1.5, -1, -0.5, 0, 1, 2])

plt.show() # affiche la figure
```

**Pour des axes présentés de manière plus familière :**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1)
plt.plot(x, y2)

# Récupération des axes
ax = plt.gca()

# Limites des axes
ax.set_xlim(-6.5, 6.5) 
ax.set_ylim(-1.8, 2)

# Ajout d'une grille personnalisable
ax.grid( visible=True, linestyle=":", linewidth=0.8, alpha=0.7 )

# Graduations sous forme de listes ou de tableaux numpy de dimension 1
ax.set_xticks(np.arange(-6, 7, 2)) 
ax.set_yticks(np.arange(-1.5, 2, 0.5))


# Axes centrés :
## 2 sont rendus invisibles
ax.spines["right"].set_visible(False) 
ax.spines["top"].set_visible(False)
## 2 sont centrés au niveau de l'origine
ax.spines["left"].set_position("zero") 
ax.spines["bottom"].set_position("zero")

plt.show() # affiche la figure
```

**Pour un repère orthonormé :**

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = np.sin(x)

plt.plot(x, y1)
plt.plot(x, y2)

# Récupération des axes
ax = plt.gca()

# Axes centrés :
## 2 sont rendus invisibles
ax.spines["right"].set_visible(False) 
ax.spines["top"].set_visible(False)
## 2 sont centrés au niveau de l'origine
ax.spines["left"].set_position("zero") 
ax.spines["bottom"].set_position("zero")

# Repère orthonormé
ax.set_aspect("equal")

plt.show() # affiche la figure
```

## Expression plus robuste/Figures multiples

### Une figure unique, un graphique unique

Pour être plus rigoureux et pour pouvoir traiter des graphiques plus complexes, on explicite la figure (c'est-à-dire l'espace de la représentation graphique) et les axes (ensemble de tous les éléments d'un cadre graphique, dont le repère, les courbes, ...).

On va les définir par la commande `fig, ax = plt.subplots()`.

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)
y1 = np.cos(x)
y2 = 2* np.sin(x)

fig, ax = plt.subplots()

ax.plot(x, y1)
ax.plot(x, y2)

# Axes centrés :
## 2 sont rendus invisibles
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)
## 2 sont centrés au niveau de l'origine
ax.spines["left"].set_position("zero")
ax.spines["bottom"].set_position("zero")

# Repère orthonormé
ax.set_aspect("equal")

plt.show() # affiche la figure
```

**Remarques importantes :** 
- On peut définir la taille de la figure par l'instruction `fig, ax = plt.subplots(figsize=(L, h))`, $L$ et $h$ étant exprimés en pouces. `fig, ax = plt.subplots(figsize=(L, h),dpi=...)` permet de définir la résolution en piwels par pource.
- `fig` est défini mais non utilisé au niveau de l'exemple précédent. Il apparaît :
    - lorsque l'on sauvegarde la figure : `fig.savefig("nom_fichier..png", dpi=...)`
    - lorsque l'on définit un titre global à la figure : `fig.suptitle("Titre")`
    - lorsque l'on souhaite adapter la disposition de plusieurs graphique dans une même figure : `fig.tight_layout()`

### Plusieurs figures distinctes

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)

fig1, ax1 = plt.subplots()
ax1.plot(x, np.cos(x))
ax1.set_title("Cosinus")

plt.show()

fig2, ax2 = plt.subplots()
ax2.plot(x, np.sin(x))
ax2.set_title("Sinus")

plt.show()
```

### Une figure unique, plusieurs graphiques

- Syntaxe de base : `plt.subplots(nbre_lignes, nbrecolonnes)`
- Pour des grilles irrégulières : `add_gridspec` ou `subplot_mosaic`

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))  # 2 lignes, 2 colonnes

axes[0, 0].plot(x, np.cos(x))
axes[0, 0].set_title("Cosinus")

axes[0, 1].plot(x, np.sin(x))
axes[0, 1].set_title("Sinus")

axes[1, 0].plot(x, x**2)
axes[1, 0].set_title("Carré")

axes[1, 1].plot(x, x**3)
axes[1, 1].set_title("Cube")

fig.tight_layout()  # évite que titres/labels se chevauchent
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)

fig, axes = plt.subplot_mosaic(
    """
    AAB
    CDB
    """,
    figsize=(10, 6)
)

axes["A"].plot(x, np.cos(x))
axes["A"].set_title("A : Cosinus (2 cases)")

axes["B"].plot(x, np.sin(x))
axes["B"].set_title("B : Sinus (colonne entière)")

axes["C"].plot(x, x**2)
axes["C"].set_title("C : Carré")

axes["D"].plot(x, x**3)
axes["D"].set_title("D : Cube")

fig.tight_layout()
plt.show()
```

```python
fig = plt.figure(figsize=(10, 8))
grille = fig.add_gridspec(2, 2)

ax1 = fig.add_subplot(grille[0, :])    # toute la première ligne
ax2 = fig.add_subplot(grille[1, 0])    # deuxième ligne, colonne 0
ax3 = fig.add_subplot(grille[1, 1])    # deuxième ligne, colonne 1

ax1.plot(x, np.cos(x))
ax1.set_title("Graphique horizontal")

ax2.plot(x, np.sin(x))
ax3.plot(x, x**2)

fig.tight_layout()
plt.show()
```

**Remarque :** Pour partager les mêmes unités sur plusieurs graphiques, lors de la définition de la figure et des axes, on précise les attributs `sharex=True, sharey=True`

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(-5, 5, 101)

fig, axes = plt.subplots(2, 2, sharex=True, sharey=True, figsize=(10, 8))  # 2 lignes, 2 colonnes

axes[0, 0].plot(x, np.cos(x))
axes[0, 0].set_title("Cosinus")

axes[0, 1].plot(x, np.sin(x))
axes[0, 1].set_title("Sinus")

axes[1, 0].plot(x, x**2)
axes[1, 0].set_title("Carré")

axes[1, 1].plot(x, x**3)
axes[1, 1].set_title("Cube")

fig.tight_layout()  # évite que titres/labels se chevauchent
plt.show()
```

## Autres graphiques

### Diagramme en barres

- barres verticales : `bar` 
- barres horizontales : `barh`

```python
import matplotlib.pyplot as plt

categories = ["Lundi", "Mardi", "Mercredi", "Jeudi"]
valeurs = [12, 19, 7, 15]
couleurs = ["skyblue", "salmon", "gold", "lightgreen"]

fig, ax = plt.subplots()

ax.bar(categories, valeurs, color=couleurs)
ax.set_title("Nombre de visiteurs par jour")

plt.show()
```

### Diagramme circulaire

- fonction `pie` 
- argument `autopct` pour afficher les pourcentages

```python
import matplotlib.pyplot as plt


categories = ["Lundi", "Mardi", "Mercredi", "Jeudi"]
valeurs = [12, 19, 7, 15]
couleurs = ["skyblue", "salmon", "gold", "lightgreen"]

fig, ax = plt.subplots()

ax.pie(valeurs, labels=categories, autopct="%1.1f%%",  colors=couleurs)
ax.set_title("Répartition des visiteurs par jour")

plt.show()
```

### Histogramme

- fonction `hist`
- argument `bins` pour fixer le nombre de classes

```python
import matplotlib.pyplot as plt

notes = [8, 12, 15, 9, 11, 13, 14, 10, 16, 12, 13, 9, 17, 11, 10]

fig, ax = plt.subplots()

ax.hist(notes, bins=8, color="orange", edgecolor="black")
ax.set_xlabel("Note")
ax.set_ylabel("Effectif")
ax.set_title("Répartition des notes")

plt.show()
```

```python
import matplotlib.pyplot as plt

notes = [8, 12, 15, 9, 11, 13, 14, 10, 16, 12, 13, 9, 17, 11, 10]

fig, ax = plt.subplots()

effectifs, bornes, barres  = ax.hist(notes, bins=8, edgecolor="black")

couleurs = ["red", "orange", "gold", "green", "teal", "blue", "purple", "pink"]

for barre, couleur in zip(barres , couleurs):
    barre.set_facecolor(couleur)

ax.set_title("Répartition des notes")

plt.show()
```

**Dégradé de couleurs selon la hauteur :** avec de nombreux dégradés automatiques :
- `cm.magma`
- `cm.plasma`
- `cm.viridis`
- `cm.inferno`
- `cm.Blues`
- `cm.Oranges`
- `cm.Reds`
- `cm.Greys`
- ...

```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm

notes = [8, 12, 15, 9, 11, 13, 14, 10, 16, 12, 13, 9, 17, 11, 10]

fig, ax = plt.subplots()

effectifs, bornes, barres = ax.hist(notes, bins=8, edgecolor="black")

# Colore chaque barre selon sa hauteur (dégradé)
for effectif, barre in zip(effectifs, barres):
    barre.set_facecolor(cm.Oranges(effectif / max(effectifs)))

ax.set_title("Répartition des notes")

plt.show()
```

### Heatmap

- fonctions  `imshow` et `colorbar`
- argument `cmap` pour définir la palette de couleurs ("viridis", "hot", "coolwarm"...). 
- `fig.colorbar()` pour associer couleurs et valeurs.

```python
import matplotlib.pyplot as plt
import numpy as np

matrice = np.random.rand(6, 6)  # matrice 6x6 de valeurs aléatoires entre 0 et 1

fig, ax = plt.subplots()

im = ax.imshow(matrice, cmap="viridis")
ax.set_title("Carte de chaleur")

fig.colorbar(im, ax=ax)  # ajoute l'échelle de couleurs à côté

plt.show()
```
