## Type d'un objet Python

Chaque objet Python admet un **type** et une **valeur**.  En Python, le type peut être **entier**, **flottant**, **liste**, **chaîne de caractères**, **booléen**, ...

Par exemple, l'objet `"Bonjour !"` est une **chaîne de caractères** de type `str` *(character string)* et de valeur `Bonjour !`.

<div style="height: 1em;"></div>

Le type d’un objet indique comment Python doit interpréter les bits codant cet objet et quelles opérations, méthodes ou attributs peuvent lui être appliqués.

Par exemple, la méthode `upper` appliquée à une chaîne de caractères produira une deuxième chaîne de
caractères obtenue à partir de la première en remplaçant chaque caractère en minuscule par le
même caractère en majuscule. Ainsi, `"Bonjour !".upper()` produira la chaîne de caractères "BONJOUR !".

Le type d'un objet Python s'obtient via la fonction `type`.


```python
type("Bonjour !")
```

```python
type(-3)
```

```python
type(23.0)
```

```python
type(True)
```

```python
type([1,5,-2,0])
```

## Affectation, variable

Une **variable** permet de lier un objet Python à une "étiquette", de le référencer. L'**affectation** d'un objet à une variable se fait via l'utilisation de `=`.

```python
a = "Au revoir"
print(a.lower())
```

L'instruction `a = "Au revoir"` :
- crée l'objet de type chaîne de caractères et de valeur "Au revoir" ;
- affecte cet objet à la variable `a`.

L'instruction `print(a.lower())` :
- crée un objet de type chaîne de caractères par application de la méthode `lower` à l'objet associé à la variable `a` ;
- affiche cet objet.

L'objet initial n'est pas modifié par la méthode `lower`.

**Remarque :** Il faut faire attention, lorsque l'on applique une méthode à un objet, à ce que celle-ci soit compatible avec le type de l'objet.

```python
b = 1
b.upper()
```

### Affectation simultanée

Pour affecter simultanément des valeurs à plusieurs variables, on utilise le séparateur `,`.

```python
a, b ,c = 1, 3, 8
print("a = ",a)
print("b = ",b)
print("c = ",c)
```

**Remarque :** L'égalité `a, b ,c = 1, 3, 8` correspond en fait à une égalité d'objets Python dénommés **tuples**.

   ## `int`, `float` et `eval`

Lorsque l'on demande à l'utilisateur d'entrer une valeur numérique via `input`, la valeur récupérée sera une chaîne de caractères et devra être convertie en entier ou décimal.

```python
a = input("Entrer un entier :")
print(type(a))
b = int(a)%2
print(b)
```

Il également possible d'utiliser la fonction `eval` beaucoup plus flexible au niveau du typage et qui permet des conversions plus complexes. La fonction `eval` convertit une chaîne de caractère en **expression Python**.

```python
a = "3.0"
print(eval(a)+2)
```

```python
b = "2 + (5-4)*3"
print(eval(b))
```

Le problème principal de la fonction `eval` est qu'elle convertit et interprète une chaîne de caractères en code Python ouvrant alors toutes les possibilités à l'utilisateur. En conclusion, **IL NE FAUT PAS UTILISER `eval` avec `input` sans sécurité !**
