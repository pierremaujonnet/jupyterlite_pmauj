# Chaînes de caractères

## Création d'une chaîne de caractères

### Sur une ligne 

Pour créer une chaîne de caractères sur une seule ligne, on peut utiliser les guillemets `""` ou `''`, en particulier pour écrire des guillemets dans une chaîne de caractères.

```python
c1 = "Bonjour !"
c2 = 'Au revoir !'
c3 = 'Je te dis "Bonjour !"'

print(c1)
print(c2)
print(c3)


# Création d'une chaîne de caractères vide
c4 = ""
```

### Sur plusieurs lignes

Pour créer une chaîne de caractère sur plusieurs lignes, on utilise des triples guillemets `""" """` ou `''' '''`.

```python
c5 = """Je souhaite écrire
un texte sur plusieurs lignes.
A bientôt !"""

print(c5)
```

### f-strings

Pour *insérer les valeurs de variables* à l'intérieur d'une chaîne de caractères facilement, on peut utiliser une **chaîne de caractères formatée** notée `f""` ou `f''` en indiquant les variables entre accolades `{}`.

```python
a = 5 
prenom = "Bob"

c6 = f"Bonjour {prenom}. Tu viens dans {a+10} minutes ?"

print(c6)
```

**Remarque importante :** Il est possible d'utiliser des règles de formatage spécifiques pour les nombres et les dates.

```python
n = 159451500

# Ecriture des entiers
print(f"{n}")
print(f"{n:,}")
print(f"{n:+}") 

# Changement de base
print(f"{n:b}")
print(f"{n:x}")
print(f"{n:X}")
```

```python
x = 351.21548607

# Ecriture des décimaux
print(f"{x:.5e}")
print(f"{x:.2f}")
print(f"{x:.5f}")
print(f"{x:.2g}")
print(f"{x:.5g}")
print(f"{x:g}")
```

```python
p = 0.458578

# Pourcentages
print(f"{p:%}")
print(f"{p:.1%}")
```

```python
from datetime import datetime

# import locale
#  locale.setlocale(locale.LC_TIME, "fr_FR.UTF-8"
    
date1 = datetime(2026, 7, 14, 9, 30)
date2 = datetime.now()

print(f"{date1:%d/%m/%Y %H:%M}")
print(f"{date2:%A %d %B %Y - %H:%M}")
```

### Echappement des caractères - Caractères spéciaux

Pour utiliser certains caractères particuliers, on peut utiliser le symbole d'échappement `\`.
- `"\n"` : retour à la ligne
- `"\t"` : tabulation
- `"\\"` : antislash 
- `"\""` : guillemet double 
- `"\'"` : apostrophe
- ...

```python
c7 = "Bonjour.\nComment allez-vous ?\n\t- Très bien merci."

print(c7)
```

**Remarque :** Pour éviter l'interprétation des chaînes de caracètres, on peut utiliser des **chaînes brutes/raw strings** notées `r""` ou `r''`.

```python
c8 = r"Bonjour.\nComment allez-vous ?\n\t- Très bien merci."

print(c8)
```

## Indices et slicing

Si `c` est une chaîne de caractère, on peut extraire un caractère `c`.
- `c[0]` : **premier** élément
- `c[i]` : élément d'**indice i**
- `c[-1]` : **dernier** élément
Pour extraire un ensemble de caractères espacé régulièrement, on peut utiliser le **slicing** `c[debut:fin:pas]`.

```python
chaine = "Exemple de texte comportant un nombre suffisant de caractères."

print(chaine[0])
print(chaine[2])
print(chaine[-1])
print(chaine[2:6])
print(chaine[9:])
print(chaine[:12])
print(chaine[2:15:2])
print(chaine[::-1])
print(chaine[14:1:-2])
```

**Remarque :** Une chaîne de caractères est un objet **immuable**, ce qui signifie que l'on ne peut pas changer cette chaîne de caractères *en place*.

```python
chaine = "Exemple de texte comportant un nombre suffisant de caractères."

chaine[2] = "e"
```

## Opérations et fonctions spécifiques

### Concaténation, répétition, nombre de caractères

- `c1 + c2` : **concaténation** des chaînes `c1` et `c2`
- `n*c` : **répétition** $n$ fois de la chaîne `c`
- `len(c)` : **nombre de caractères**

```python
c9 = "Bonjour"
c10 = "Au revoir"

print(c9 + " - " + c10)
print(7*c9)
print(len(c9))
```

### Méthodes usuelles

- `chaine.upper()` : caractères en majuscules
- `chaine.lower()` : caractères en minuscules
- `chaine.capitalize()` : première lettre en majuscule, les autres en minuscules

```python
c11 = "BoNJour !"

print(c11.lower())
print(c11.upper())
print(c11.capitalize())
```

- `chaine.strip()` : retire les espaces des deux côtés
- `chaine.strip(c)` : retire les caractères des deux côtés
- `chaine.lstrip()`/`chaine.rstrip()` : retire les espaces à gauche/à droite

```python
c12 = "   -----Bonjour- - ----"

print(c12.strip())
print(c12.strip("-"))
print(c12.strip(" -"))
```

- `c in chaine` : `True` si la sous-chaîne `c` est dans la chaîne, `False` sinon
- `chaine.count(c)` : compte le nombre d'occurrences de la sous-chaîne `c`
- `chaine.find(c)` : indice de la première occurrence de la sous-chaîne `c`, $-1$ si absent
- `chaine.index(c)` : indice de la première occurrence de la sous-chaîne `c`, erreur si absent

```python
c13 ="Exemple de texte comportant un nombre suffisant de caractères."

print("Exemple" in c13)
print("exemple" in c13)
print(c13.count("e"))
print(c13.count("om"))
print(c13.find("tex"))
print(c13.find("cor"))
print(c13.index("tex"))
print(c13.index("cor"))
```

- `chaine.split()` : découpage d'une chaîne de caractère dans une liste selon les espaces
- `chaine.split(sep)` : découpage d'une chaîne de caractère dans une liste avec séparateur indiqué
- `chaine.replace(ancien, nouveau)` : remplacement d'éléments d'une chaîne
- `c.join(liste)` : assemblage des éléments d'une liste

```python
c14 = "Bonjour à vous."
c15 = "Alban ; Léane ; John"

print(c14.split())
print(c15.split(" ; "))
print(c14.replace("o","a"))
print(c15.replace(" ",""))

liste = ["un", "deux", "trois", "quatre"]
print(" - ".join(liste))
```

**Remarque importante :** `replace` ne change pas la chaîne initiale mais en crée une nouvelle qu'il faudra réaffecter.

- `chaine.startswith(c)` : `True` si la sous-chaîne `c` est située au début de la chaine, `False` sinon
- `chaine.endswith(c)` : `True` si la sous-chaîne `c` est située à la fin de la chaine, `False` sinon
- `chaine.isdigit()` : `True` si la chaîne est composée uniquement de chiffres, `False` sinon
- `chaine.isalpha()` : `True` si la chaîne est composée uniquement de lettres, `False` sinon
- ...
