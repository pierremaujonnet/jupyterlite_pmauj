# Fonctions en Python

## Créer une fonction en Python

En Python, une fonction permet de découper et de structurer le code en parties simples qui pourront être réutilisées plusieurs fois. C'est un objet Python qui contient un certain nombre d'instructions exécutées à l'appel de la fonction.

Une fonction est définie à l'aide du mot-clé `def`. Le nom de la fonction *(insécable)* est suivi de parenthèses `()` *(tout comme lors de son appel)* et de deux points `:`. L'espace de définition est réservé par une indentation.

```python
def premiere_fonction():
    """
    Début des instructions de la fonction  Afficher "Bonjour"
    
    """
    print("Bonjour !")

# Fin de la définition de la fonction marquée par un retour d'indentation
premiere_fonction()
```

La définition d'une fonction peut contenir un ou plusieurs paramètres.

```python
def affichage(texte):
    texte_final = texte.upper()
    print(texte_final)
    
affichage("Bonjour")
```

```python
def affichage2(texte1, texte2):
    texte_final = texte1.upper() + " " + texte2.lower()
    print(texte_final)
    
affichage2("Bonjour", "Au revoir")
    
```

## Utilisation du mot-clé `return` 

Le mot-clé `return` permet de **retourner une ou plusieurs valeurs**. Ce mot-clé a pour effet de **sortir du bloc fonction** immédiatement.

```python
def fonction_return():
    print("Bonjour")
    return "Fin"

print(fonction_return())
```

```python
def fonction_return2():
    print("Bonjour")
    return "Fin"
    print("Au revoir")

print(fonction_return2())
```

```python
def fonction_return3():
    salut = "Bonjour"
    prenom = "Mathilde"
    return salut, prenom

var1, var2 = fonction_return3()

print(var1 + " " + var2 + " !")
```

```python
def fonction_return4():
    a = 3
    b = 5
    somme = a + b
    produit = a * b
    return somme, produit

resultats = fonction_return4()

print(resultats[0])
print(resultats[1])

type(fonction_return4())
```

**Remarque :** Lorsque la fonction retourne plusieurs valeurs, elle le fait sous forme d'un **tuple**.

## Utilisation d'instructions tierces

### Importation de modules

Il est fréquent d'avoir à utiliser dans des fonctions des objets Python définis dans des modules supplémentaires. Il est alors nécessaire d'importer ces modules.

```python
from math import sqrt

def distance(xA, yA, xB, yB):
    return sqrt((xB-xA)**2+(yB-yA)**2)

distance(1, 3, 4, 7)
```

###  Instructions conditionnelles et boucles

```python
def maximum(a, b):
    if a >= b:
        return a
    else:
        return b
    
maximum(4, 6)
```

```python
def somme():
    n = 1
    resultat = 0
    while 1/n > 0.003:
        resultat = resultat + 1/n # Peut s'écrire également : resultat += 1/n
        n = n +1 # Peut s'écrire également : n += 1
    return resultat

print(somme())
```

## Paramètres présentant des valeurs par défaut

Il est possible de donner des valeurs par défaut à certains paramètres. Lors de l'appel de la fonction, il n'est pas nécessaire de leur donner une valeur (ils prendront alors leur valeur par défaut).

Lors de la définition de la fonction, les paramètres présentant des valeurs par défaut doivent être situés **après** les paramètres n'ayant pas de valeur par défaut.

```python
def dire_message(nom, prenom, message = "Bonjour"):
    return message + " " + prenom + " " + nom + "!"

print(dire_message("Durand", "Kevin"))
print(dire_message("Durand", "Kevin","Au revoir"))
```

## Lorsque le nombre de paramètres n'est pas connu

Lorsque le nombre de paramètres n'est pas connu a priori, on définit comme paramètre un **tuple** dont le nom est précédé d'une `*`. Celui-ci doit être placé après les éventuels paramètres obligatoires.

```python
def affiche_notes(*notes):
    for note in notes:
        print(note)

affiche_notes(9, 10, 8.5)
```

```python
def moyenne(prenom,*notes):
    result = 0
    for note in notes:
        result += note
    result /= len(notes)
    print(f"La moyenne de {prenom} est {result}.")

moyenne("Lucie", 13, 14, 20, 9, 10)
```

**Remarque :** On a utilisé ici une chaîne formatée/*f-string* qui permet d'insérer facilement les valeurs de variables dans une chaîne de caractère.

Il peut être utile de convertir les entrées d'arguments nommés d'une fonction en un **dictionnaire**. Le nom du paramètre sera alors précédé de `**`.

```python
def afficher(**informations):
    print(informations)
    
afficher(nom = "Durand", prenom = "Jonathan", ville = "Paris")
```

```python
def moyenne_ponderee(**notes):
    somme = 0
    poids = 0
    for matiere, note in notes.items():
        if matiere == "maths":
            somme += 5*note
            poids += 5
        elif matiere == "physique":
            somme += 4*note
            poids += 4
        else:
            somme += 2*note
            poids += 2
    return somme/poids

print(moyenne_ponderee(maths=16, physique=14, anglais=12))
```

## Fonctions `lambda`

Une fonction `lambda` est une fonction à l'**écriture simplifiée** et n'ayant **pas nécessité d'être dénommée**. Ainsi, une telle fonction présente un nombre limité d'instructions.

```python
g = lambda x : x**3

print(g(10))
```

Un des intérêts des fonctions lambda est lorsqu'un des paramètres d'une fonction est une fonction.

```python
def taux_accroissement(f,a,b):
    return (f(b)-f(a))/(b-a)

print(taux_accroissement(lambda x : x**2, 0, 3))
```
