# Listes en Python

## Création d'une liste - Accès aux éléments

### Création 

En Python, une liste permet de stocker une suite de valeurs qui peuvent être de types distincts.

```python
# Création d'une liste vide
liste1 = []

# Liste contenant des nombres
liste2 = [1, 2, 4, 8, 16]

# Liste contenant des objets divers
liste3 = [True, "Bonjour", 3.2, -5, "Ok"]
```

### Utilisation de range

On peut obtenir facilement une liste d'entiers espacés régulièrement à l'aide de la fonction `range`.

```python
liste4 = list(range(3,10))

print(liste4)
```

### Compréhension de liste

Un liste peut être créée **par compréhension de liste**.

```python
liste1 = [x**2 for x in range(10)]
print(liste1)

liste2 = [x-5 for x in liste1]
print(liste2)

liste3 = [n for n in range(-5, 10) if n%3 == 0] # Liste des entiers divisibles par 3, compris entre -5 inclus et 10 exclu
print(liste3)
```

### Accès d'un élément - Modification

Pour accéder (ou modifier) à un élément d'une liste, on utilise la notation `liste[indice]`. 
- L'indice du **premier élément** de la liste est 0.
- L'indice du **dernier élément** de la liste est -1.

```python
liste = [-3, 2, -8, 0, 1, 2, 1, -3.6]

print(liste[0]) # Affiche le premier élément de la liste
print(liste[3]) # Affiche le quatrième élément de la liste
print(liste[-1]) # Affiche le dernier élément de la liste

liste[2] = -5 # Redéfinit le troisième élément de la liste
print(liste)
```

## Ajout et suppression d'éléments

### Ajout d'éléments

- `liste.append(element)` : ajout d'un élément **à la fin** de la liste
- `liste.insert(indice, element)` : ajout d'un élément **au niveau d'un indice donné** dans la liste
- `liste.extend(liste2)` : ajout de tous les éléments d'une deuxième liste **à la fin** de la liste initiale
- `liste1 + liste2` : nouvelle liste obtenue par **concaténation** de deux listes sans modifier aucune des deux listes

```python
l1 = [1, 2, 3]
l2 = [4, 5, 6, 7]

l1.insert(2, -5)
print(l1)

l1.append(-6)
print(l1)

l1.extend(l2)
print(l1)

print(l1+l2)
```

### Suppression d'éléments

- `del liste[indice]` : supprime un élément **d'indice donné**
- `liste.remove(element)` : supprime **la première occurrence** d'un objet donné
- `liste.pop()` : supprime *et récupère* **le dernier élément** de la liste
- `liste.pop(indice)` : supprime *et récupère* l'élément **d'indice donné** de la liste

```python
liste = [2, 3, -1, 1, 2, 4, 3, 5, 3]

liste.remove(3)
print(liste)

del liste[2]
print(liste)

print(liste.pop())
print(liste)

print(liste.pop(1))
print(liste)
```

## Compléments importants

### Fonctions applicables aux listes

- `len(liste)` : **nombre d'éléments** de la liste
- `sum(liste)` : somme des éléments de la liste
- `max(liste)` : plus grand élément de la liste 
- `min(liste)` : plus petit élément de la liste 
- `liste.index(element)` : **indice de la première occurrence** de l'élement dans la liste
- `liste.count(element)` : donne **le nombre d’occurrences** de l'élément dans la liste
- `liste.reverse()` : **inverse l’ordre** des éléments de la liste
- `liste.sort()` : **réordonne** les éléments de la liste *dans la liste*
- `sorted(liste)` : crée une *nouvelle liste* à partir des éléments **triés** de la liste initiale

```python
liste = [1, -3, 5, -7, 12, 5, -3, 1, 1, 6, -8, 1, 7]

print(len(liste))
print(sum(liste))
print(max(liste))
print(min(liste))
print(liste.count(1))
print(liste.index(-3))
```

```python
liste = [1, -3, 5, -7, 12, 5, -3, 1, 1, 6, -8, 1, 7]

liste.reverse()
print(liste)
print(sorted(liste))
print(liste)
liste.sort()
print(liste)
```

### Slicing

Pour extraire une partie d'une liste, on peut utiliser le **slicing**.

```python
liste = [0, 1, 2, 3, 4, 5, 6, 7]

print(liste[:])
print(liste[2:])
print(liste[:5])
print(liste[1:6])
print(liste[1:7:2])
```

### Remarque importante

De nombreux attributs et fonctions applicables aux listes peuvent être transposés aux chaînes de caractères.

```python
c = "Je travaille le langage Python"

print(len(c))
print(c[3])
print(c[:6])
print(c[2:12:3])
print(max(c))
print(sorted(c))
print(c.count("a"))
print(c.count("le"))
print(c.index("le"))
```
