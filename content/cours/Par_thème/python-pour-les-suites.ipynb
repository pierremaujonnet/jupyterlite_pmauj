# Calculs usuels

## Calcul d'un terme particulier

### Suites définies explicitement

Considérons la suite $(u_n)$ définie par $u_n=2n^2+1$.

```python
def suite_u(n):
    return 2*n**2+1

# Exemple d'utilisation
print(suite_u(5)) # Affiche u_5
```

### Suites définies par récurrence

Considérons la suite $(v_n)$ définie par $\begin{cases}
					v_0=1\\v_{n+1}=6v_n-3\\
				\end{cases}$.

```python
def suite_v(n):
    v = 1 # Terme initial
    for i in range(n):
        v = 6*v-3
    return v

# Exemple d'utilisation
print(suite_v(10)) # Affiche v_10
```

### Variantes (suites définies par récurrence)

```python
def v_variante_while(n):
    v = 1  # Terme initial
    i = 0
    while i < n:
        v = 6 * v - 3
        i = i + 1 # écriture équivalente : i += 1	 
    return v

def v_variante_recursivite(n):
    if n == 0:
        return 1  # Condition de base : v_0 = 1
    else:
        return 6 * v_variante_recursivite(n - 1) - 3  # Formule de récurrence
```

## Déterminer le rang pour dépasser un seuil

Soit $(w_n)$ la suite définie par $\begin{cases}
			w_0=100\\w_{n+1}=0,8w_n
		\end{cases}$. Déterminer le plus petit entier $n_0$ tel que $w_{n_0}<0,01$.

```python
def rang_seuil():
    n = 0
    w = 100
    while w >= 0.01:
        w = 0.8*w
        n = n+1
    return n
```

## Somme de termes consécutifs jusqu'au terme de rang $n$

Plusieurs solutions sont possibles.

```python
def S(n):
    w = 100 
    somme = w
    for _ in range(n):
        w = 0.8 * w
        somme = somme+w
    return somme
```

```python
def S_rec(n):
    if n == 0:
        return 100
    else:
        return (100*(0.8)**n)+S_rec(n - 1)
```

```python
def S_while(n):
    w = 100
    somme = w
    compteur = 0
    while compteur < n:
        w = 0.8 * w
        somme += w
        compteur += 1
    return somme
```

# Liste de termes/Représentation graphique

## Liste de termes consécutifs

### Suites définies explicitement : méthode par compréhension de liste

```python
def liste_u(n):
    return [2 * i**2 + 1 for i in range(n+1)] 

# Exemple d'utilisation
print(liste_u(5))
```

### Suites définies par récurrence : méthode par ajouts successifs

```python
def liste_v(n):
    v = 1
    liste = [v]
    for i in range(n):
        v = 6*v-3
        liste.append(v)
    return liste

# Exemple d'utilisation
print(liste_v(10))
```

### Variante

```python
def liste2_v(n):
    liste = [1]
    for i in range(n):
        liste.append(6*liste[-1]-3)
    return liste
```

## Représentation graphique

### Nuage de points

```python
import matplotlib.pyplot as plt
			
def tracer_v(n):
    plt.clf() # facultatif
    x = range(n + 1)
    y = liste_v(n)
    plt.scatter(x, y, marker='o', color ='cadetblue')
    plt.xlabel("rang n")
    plt.ylabel("valeur v_n")
    plt.title("Représentation de la suite (v_n)")
    plt.grid()
    plt.show()

# Exemple d'utilisation
tracer_v(7)
```
