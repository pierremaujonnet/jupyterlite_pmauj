## Géométrie analytique
### Distance entre deux points
Soient $A$ et $B$ deux points de coordonnées respectives $(x_A,y_A)$ et $(x_B,y_B)$ dans un repère **orthonormé**. Déterminer la distance $AB$.

```python
# Importation de la fonction racine carrée
from math import sqrt

def distance(xA, yA, xB, yB):
    return sqrt((xB - xA) ** 2 + (yB - yA) ** 2)

distance(1, 2, -4, 6)
```

### Milieu d'un segment
Soient $A$ et $B$ deux points de coordonnées respectives $(x_A,y_A)$ et $(x_B,y_B)$ dans un repère. Déterminer les coordonnées du milieu $I$ du segment $[AB]$.

```python
def coordonneesMilieu(xA, yA, xB, yB):
    return (xA + xB) / 2, (yA + yB) / 2

coordonneesMilieu(1, 2, -4, 6)
```

### Coefficient directeur et ordonnée à l'origine
Soient $A$ et $B$ deux points distincts de coordonnées respectives $(x_A,y_A)$ et $(x_B,y_B)$ dans un repère tels que $x_A\neq x_B$.
- Déterminer le coefficient directeur de la droite $(AB)$.
- Déterminer l'ordonnée à l'origine de la droite $(AB)$.

```python
def coefficientDirecteur(xA, yA, xB, yB):
    if xB == xA:        
        return "La droite est verticale."
    else:
        return (yB - yA) / (xB - xA)

def ordonneeOrigine(xA, yA, xB, yB):
    if xB == xA:
        return "La droite est verticale."
    else:
        m = coefficientDirecteur(xA, yA, xB, yB)
        return yA - m * xA

print(coefficientDirecteur(1, 2, -4, 6))
print(ordonneeOrigine(1, 2, -4, 6))
```

### Alignement de trois points
Soient $A$, $B$ et $C$ trois points distincts de coordonnées respectives $(x_A,y_A)$, $(x_B,y_B)$ et $(x_C,y_C)$ dans un repère. Déterminer si les points $A$, $B$ et $C$ sont alignés.

```python
def alignement(xA, yA, xB, yB, xC, yC):
    x_AB = xB - xA
    y_AB = yB - yA
    x_AC = xC - xA
    y_AC = yC - yA
    determinant = x_AB * y_AC - y_AB * x_AC
    if determinant == 0:
        return "Les points A, B et C sont alignés."
    else:
        return "Les points A, B et C ne sont pas alignés."

print(alignement(1, 2, -4, 6, 11, -6))
```

### Appartenance d'un point à une courbe
Soit $f$ une fonction de courbe représentative $\mathcal{C}$ dans un repère donné.
Soit $A(x_A,y_A)$ un point tel que $x_A$ soit dans l'ensemble de définition de $f$. Déterminer si $A$ appartient à $\mathcal{C}$.

```python
def f(x):
    return - x**3 + 5*x + 3

def appartient(f, x_A, y_A):
    return f(x_A) == y_A

appartient(f, 1, 7)
```

**Remarque importante**

Dans le cas de calculs comprenant des valeurs non entières, on pourra remplacer `==` (qui pourra donner de faux cas négatifs) par `isclose`.

```python
from math import isclose

print(0.1 + 0.2 == 0.3)
print(isclose(0.1 + 0.2, 0.3))
```

## Fonctions
### Déterminer une valeur approchée d'un extremum par balayage

```python
def maximum_balayage(f, a, b, pas):
    x_max = a
    maximum = f(a)
    x = a + pas
    while x <= b:
        y = f(x)
        if y > maximum:
            maximum = y
            x_max = x
        x += pas
    return x_max, maximum

def f(x):
    return - 2*x**2 + 7*x + 3

maximum_balayage(f, -5, 5, 0.001)
```

### Valeur approchée de $\sqrt{2}$ par balayage

```python
def racine2(pas):
    x = 0
    while x*x < 2:
        x += pas
    return x

racine2(0.001)
```

## Aléatoire
### Simuler un lancer de dé
Simuler un lancer de dé cubique équilibré.

```python
# Importer la fonction randint du module random
# Cette fonction permet de générer un entier aléatoire compris entre deux bornes (incluses).
from random import randint

def lancer_de():
    return randint(1, 6)

lancer_de()
```

### Simuler plusieurs lancers de dé
Simuler $n$ lancers d'un dé cubique équilibré. Déterminer la fréquence d'apparition d'une face $f$.

```python
from random import randint

def frequence_lancers_de(n, f):
    nombre_apparitions = 0
    for _ in range(n):
        if lancer_de() == f:
            nombre_apparitions += 1
    return nombre_apparitions/n

frequence_lancers_de(1000, 4)
```

Simuler $n$ lancers d'un dé cubique équilibré et renvoyer les résultats sous forme d'une liste

```python
from random import randint

def lancers_de(n):
    liste_resultats = []
    for _ in range(n):
        liste_resultats.append(randint(1, 6))
    return liste_resultats

lancers_de(10)
```

Variante

```python
from random import randint

def lancers_de(n):
    return [randint(1, 6) for _ in range(n)]

lancers_de(10)
```

## Arithmétique
### Divisibilité

Soient $a$ et $b$ deux entiers relatifs. Déterminer si $a$ est un multiple de $b$.

```python
def multiple(a, b):
    if b == 0:
        return a == 0
    else:
        return a % b == 0

multiple(10, 4)
```

Soient $a$ et $b$ deux entiers naturels avec $b\neq 0$. Déterminer si $a$ est un multiple de $b$ sans utiliser `%`.

```python
def multiple(a, b):
    while a >= b:
        a -= b
    return a == 0

multiple(10, 4)
```

## Divers
### Fréquence d'une lettre dans une chaîne de caractères

```python
def frequence_lettre(chaine, lettre):
    nombre = 0
    for caractere in chaine:
        if caractere == lettre:
            nombre += 1
    return nombre / len(chaine)

frequence_lettre("Bonjour à vous", "o")
```

**Remarque importante**

Pour ne compter que les lettres et regrouper minuscules et majuscules.

```python
def frequence_lettre_2(chaine, lettre):
    nombre = sum(1 for caractere in chaine
                 if caractere.isalpha()
                 and caractere.lower() == lettre.lower())

    total = sum(1 for caractere in chaine if caractere.isalpha())

    return nombre / total

frequence_lettre_2("Bonjour à vous", "b")
```
