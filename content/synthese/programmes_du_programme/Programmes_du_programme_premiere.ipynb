## Suites
### Termes des éléments d'une suite définie par récurrence
Déterminer une valeur apporchée d'un terme de la suite définie par : $$\begin{cases}u_0=3\\u_{n+1}=\dfrac{5u_n-2}{6}\end{cases}$$

**Utilisation de `for`**

```python
def u(n):
    uk = 3
    for k in range(n):
        uk = (5 * uk - 2) / 6
    return uk

u(10)
```

**Utilisation de `while` *(version équivalente au `for`précédent)***

```python
def u(n):
    uk = 3
    k = 0
    while k < n:
        uk = (5 * uk - 2) / 6
        k = k + 1
    return uk 

u(10)
```

**Utilisation de `while` (bis)

```python
def u(n):
    uk = 3
    while n > 0:
        uk = (5 * uk - 2) / 6
        n = n - 1
    return uk

u(10)
```

**Fonction récursive**

```python
def u(n):
    if n == 0:
        return 3
    else:
        return (5 * u(n - 1) -2) / 6

u(10)
```

### Liste des termes successifs d'une suite
**Suites définies explicitement**

Déterminer une liste des $n$ premiers termes de la suite définie par $$v_n=\dfrac{5+(-1)^n}{n}$$ où $n$ est un entier naturel non nul.


```python
def liste_v(n):
    liste = []
    for k in range(1,n+1):
        liste.append((5+(-1)**k)/k)
    return liste

liste_v(10)
```

**Liste par compréhension *(à privilégier)***

```python
def liste_v(n):
    return [(5+(-1)**k)/k for k in range(1,n+1)]

liste_v(10)
```

### Déterminer un seuil
**Exemple d'une suite définie explicitement**

Soit $(v_n)$ la suite définie par $$v_n=\dfrac{5+(-1)^n}{n}$$ où $n$ est un entier naturel non nul.

Déterminer le plus petit entier naturel $n$ tel que $v_n<0,0001$.

```python
def seuil_v():
    n = 1
    while (5+(-1)**n)/n >= 0.0001:
        n += 1
    return n

seuil_v()
```

**Exemple d'une suite définie par récurrence**

Soit $(u_n)$ la suite définie par  $$\begin{cases}u_0=3\\u_{n+1}=\dfrac{5u_n-2}{6}\end{cases}$$
Déterminer le plus petit entier naturel $n$ tel que $u_n+2\leq s$ où $s$ est un réel strictement positif entré par l'utilisateur.

```python
def seuil_u(s):
    n = 0
    u = 3
    while u + 2 > s:
        n += 1
        u = (5 * u - 2) / 6
    return n

seuil_u(0.0001)
```

### Suite de Fibonacci
Déterminer un terme de la suite $(F_n)$ définie par :$$\begin{cases}F_0=0\\F_1=1\\F_{n+2}=F_n+F_{n+1}\\ \end{cases}$$

```python
def fibonacci(n):
    f0 = 0
    f1 = 1
    for _ in range(n):
        f0, f1 = f1, f0 + f1
    return f0

fibonacci(10)
```

Déterminer une liste des $n+1$ premiers termes de la suite avec $n\geq 1$.

```python
def liste_fibonacci(n):
    liste = [0, 1]
    for _ in range(n-1):
        liste.append(liste[-1] + liste[-2])
    return liste

liste_fibonacci(10)
```

### Suite de Syracuse
Soit $(u_n)$ une suite telle que :
    $$\begin{cases}u_0\in\mathbb{N}^*\text{ ;}\\u_{n+1}=\dfrac{u_n}{2}\text{ si }n\text{ est pair ;}\\u_{n+1}=3u_n+1\text{ si }n\text{ est imppair.}\end{cases}$$

1. Determiner le terme $u_n$ où $n\in\mathbb{N}$.
2. Déterminer la liste des $n+1$ premiers termes.
3. Déterminer le temps de vol (premier passage à la valeur 1) selon la valeur de $u_0$.
4. Déterminer l'altitude maximale (plus grande valeur) selon la valeur de $u_0$.

```python
def u(n, u0):
    uk = u0
    for _ in range(n):
        if uk % 2 == 0:
            uk = uk // 2
        else:
            uk = 3 * uk + 1
    return uk

u(10, 15) 
```

```python
def liste_syracuse(n, u0):
    liste = [u0]
    for _ in range(n):
        if liste[-1] % 2 == 0:
            liste.append(liste[-1] // 2)
        else:
            liste.append(3 * liste[-1] + 1)
    return liste

liste_syracuse(10, 15)
```

```python
def temps_vol(u0):
    un = u0
    n = 0
    while un != 1:
        if un % 2 == 0:
            un = un // 2
        else:
            un = 3 * un + 1
        n += 1
    return n

temps_vol(15)
```

```python
def altitude_max(u0):
    un = u0
    max_altitude = u0
    n = 0
    n_max = 0
    while un != 1:
        n += 1
        if un % 2 == 0:
            un = un // 2
        else:
            un = 3 * un + 1
        if un > max_altitude:
            max_altitude = un
            n_max = n
    return max_altitude, n_max

altitude_max(15)
```

## Résolutions d'équations
### Equations du second degré
Soient $a$, $b$ et $c$ trois réels avec $a\neq0$. Déterminer les solutions éventuelles de l'équation $ax^2+bx+c=0$.

```python
def solutions(a, b, c):
    delta = b**2 - 4*a*c
    if delta < 0:
        return None
    elif delta == 0:
        return -b/(2*a)
    else:
        x1 = (-b - delta**0.5)/(2*a)
        x2 = (-b + delta**0.5)/(2*a)
        return (x1, x2)

solutions(-2, 2, 12)
```

### Algorithme de de dichotomie
Soient $a$ et $b$ deux réels avec $a<b$. Soit $f$ une fonction définie sur $[a,b]$.

On suppose que la fonction $f$ est strictement monotone sur $[a,b]$ et que l'équation $f(x)=0$ admet une solution sur l'intervalle $]a,b[$. 

Déterminer un encadrement d'amplitude au plus 0,0001 de cette solution.

```python
def dichotomie(f, a, b):
    while b - a > 0.0001:
        m = (a + b) / 2
        if f(m) == 0:
            return m
        elif f(a) * f(m) < 0:
            b = m
        else:
            a = m
    return a, b

def f(x):
    return x**2 - x - 1

%time resultat = dichotomie(f, 0, 2)
print(resultat)
```

### Méthode par balayage

```python
def balayage(f, a, b, pas):
    x = a
    fx = f(x)

    while x + pas <= b:
        y = x + pas
        fy = f(y)

        if fx * fy <= 0:
            return x, y

        x = y
        fx = fy

    return None


def f(x):
    return x**2 - x - 1

%time resultat = balayage(f, 0, 2, 0.0001)
print(resultat)
```

**Remarque importante**

Nous avons ajouté un calcul du temps d'éxécution de ces deux algortihmes pour constater la diiférence «d'efficacité» de ces algorithmes. Cette dernière dépend également de la fonction `f`.

## Fonction exponentielle
### Méthode d'Euler
Déterminer une valeur approchée de $\text{e}^x$ pour $x>0$ par la méthode d'Euler en utilisant $n$ pas.

```python
def e_Euler(x, n):
    pas = x / n
    y = 1
    for _ in range(n):
        y = y * (1 + pas)
    return y

e_Euler(1, 1000)
```

### Méthode d'Euler (représentation graphique)

```python
import matplotlib.pyplot as plt

def e_Euler_graphique(x, n):
    pas = x / n
    t = 0
    y = 1
    liste_t = [0]
    liste_y = [y]
    for _ in range(n):
        t = t + pas
        y = y * (1 + pas)
        liste_t.append(t)
        liste_y.append(y)
    plt.plot(liste_t, liste_y)
    plt.title(fr"Valeur approchée de $e^{x}$ par la méthode d'Euler")

e_Euler_graphique(1, 5)
e_Euler_graphique(1, 20)
e_Euler_graphique(1, 100)
plt.grid()
plt.show()
```

## Variables aléatoires
### Calculer les paramètres d'une loi de probabilités
Calculer l'espérance, la variance et l'écart-type d'une variable aléatoire dont la loi de probabilités est donnée ci-dessous :
$$\begin{array}{c|c|c|c|c|c|c}x_i&-3&-1&0&1&6&7\\\hline p_i&0,1&0,2&0,3&0,2&0,1&0,1\\\end{array}$$

```python
x = [-3, -1, 0, 1, 6, 7]
p = [0.1, 0.2, 0.3, 0.2, 0.1, 0.1]

def esperance(x, p):
    E = 0
    for k in range(len(x)):
        E += x[k] * p[k]
    return E

esperance(x, p)
```

```python
def variance(x, p):
    E = esperance(x, p)
    V = 0
    for k in range(len(x)):
        V += (x[k] - E)**2 * p[k]
    return V

variance(x, p)
```

```python
from math import sqrt

def ecart_type(x, p):
    return sqrt(variance(x, p))

ecart_type(x, p)
```

### Simuler une variable aléatoire
Un joueur lance un dé équilibré. 
- Si le dé tombe sur 1, le joueur gagne 1€.
- Si le dé tombe sur 6, le joueur gagne 6€.
- Dans tous les autres cas, le joueur perd 2€.

Soit $X$ la variable aléatoire associée au gain du joueur.

1. Simuler l'expérience une fois et retourner la valeur de $X$ correspondante.
2. Renvoyer dans une liste les résultats associés à $n$ répétitions de cette expérience.
3. Estimer l'espérance de $X$ par calcul de la moyenne d'un échantillon de taille $n$.
4. Estimer la variance de $X$ par calcul de la variance d'un échantillon de taille $n$.

```python
from random import randint

def jeu():
    resultat = randint(1, 6)
    if resultat == 1:
        return 1
    elif resultat == 6:
        return 6
    else:
        return -2

jeu()
```

```python
def liste_jeu(n):
    liste = []
    for _ in range(n):
        liste.append(jeu())
    return liste

liste_jeu(10)
```

```python
def moyenne_echantillon(n):
    liste = liste_jeu(n)
    return sum(liste) / n

moyenne_echantillon(10000)
```

A noter que l'espérance théorique est ici de :
$$E(X)=\dfrac{1}{6}\times 1+\dfrac{1}{6}\times6+\dfrac{2}{3}\times(-2)=-\dfrac{1}{6}\approx -0,1667$$

```python
def variance_echantillon(n):
    liste = liste_jeu(n)
    moyenne = sum(liste) / n
    variance = sum([x**2 for x in liste]) / n - moyenne ** 2
    return variance

variance_echantillon(10000)
```

La variance théorique est de : 
$$V(X)=\dfrac{1}{6}\times 1^2+\dfrac{1}{6}\times6^2+\dfrac{2}{3}\times(-2)^2-\left(-\dfrac{1}{6}\right)^2=\dfrac{317}{36}\approx 8.8056$$



Un joueur lance un dé et une pièce équilibrés simultanément.
- Si la pièce tombe sur Pile et le dé sur une face impaire, le joueur gagne 1€.
- Si la pièce tombe sur Face et le dé sur la face 2, le joueur gagne 3€.
- Si la pièce tombe sur Face et le dé sur la face 6, le joueur gagne 6€.
- Dans tous les autres cas, le joueur perd 1€.

Soit $X$ la variable aléatoire associée au gain du joueur.
Déterminer une estimation de l'espérance de $X$.

```python
from random import randint

def jeu():
    resultat_de = randint(1, 6)
    resultat_piece = randint(0, 1)
    if resultat_piece == 0:
        if resultat_de % 2 == 1:
            return 1
        else:
            return -1
    else:
        if resultat_de == 2:
            return 3
        elif resultat_de == 6:
            return 6
        else:
            return -1

def liste_jeu(n):
    liste = []
    for _ in range(n):
        liste.append(jeu())
    return liste

def esperance_jeu(n):
    liste = liste_jeu(n)
    return sum(liste) / n

esperance_jeu(10000)
```

L'espérance théorique de $X$ est :
$$E(X)=\dfrac{1}{4}\times1+\dfrac{1}{12}\times3+\dfrac{1}{12}\times6+\dfrac{7}{12}\times(-1)=\dfrac{5}{12}\approx0,4167$$
