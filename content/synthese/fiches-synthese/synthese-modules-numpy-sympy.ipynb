# Synthèses modules Numpy et Sympy

## Numpy

L'import se fait via `import numpy as np`.

### Tableaux et `dtype`

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soit A un tableau numpy, créé via <code>np.array(...)</code>.
    <ul>
        <li><code>A.dtype</code> : <strong>type numpy</strong> des éléments de A <i>(tous les éléments d'un même tableau ont le même type)</i></li>
        <li><code>A.ndim</code> : <strong>nombre de dimensions</strong> de A</li>
        <li><code>A.shape</code> : <strong>dimensions</strong> de A, sous forme de <i>tuple</i></li>
        <li><code>A.size</code> : <strong>nombre total d'éléments</strong> de A</li>
        <li><code>A.itemsize</code> : <strong>espace mémoire d'un élément</strong> de A, en octets <i>(en relation avec <code>A.dtype</code>)</i></li>
        <li><code>A.nbytes</code> : <strong>espace mémoire total</strong> des données de A, en octets</li>
    </ul>
</div>

<div style="

    <ul>
        <li><code>np.zeros(format)</code> : tableau <strong>composé uniquement de 0</strong></li>
        <li><code>np.ones(format)</code> : tableau <strong>composé uniquement de 1</strong></li>
        <li><code>np.full(format, element)</code> : tableau <strong>composé de la répétition d'un élément</strong></li>
        <li><code>np.eye(n)</code> : <strong>matrice unité</strong> d'ordre n</li>
        <li><code>np.indices(format)</code> : <strong>grilles des indices</strong></li>
        <li><code>np.arange(debut, fin, pas)</code> : tableau à 1 dimension de valeurs <strong>régulièrement espacées d'un pas fixé</strong></li>
        <li><code>np.linspace(debut, fin, nombre)</code> : tableau à 1 dimension d'<strong>un nombre fixé de valeurs régulièrement espacées</strong></li>
    </ul>
</div>

### Éléments et slicing

    padding:0.8rem 1rem;

    color:#2b2b2b;
">
Soit A un tableau numpy à 2 dimensions.
    <ul>
        <li><code>A[i,j]</code> : <strong>élément</strong> à l'intersection de la i+1-ième ligne et de la j+1-ième colonne <i>(indices numérotés à partir de 0)</i></li>
        <li><code>A[i,:]</code> : <strong>i+1-ième ligne</strong> de A</li>
        <li><code>A[:,j]</code> : <strong>j+1-ième colonne</strong> de A</li>
        <li><code>A[i,:].copy()</code>/<code>A[:,j].copy()</code> : <strong>copie</strong> de la i+1-ième ligne/j+1-ième colonne</li>
    </ul>
</div>

### Opérations

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soient A et B deux tableaux numpy de même dimension. Les opérations <code>+</code>, <code>-</code>, <code>*</code>, <code>**</code> et les fonctions usuelles s'appliquent <strong>élément par élément</strong> <i>(calcul vectorisé)</i>.
    <ul>
        <li><code>A + B</code>, <code>A - B</code>, <code>A * B</code>, <code>A ** n</code> : opérations élément par élément</li>
        <li><code>np.sqrt</code>, <code>np.exp</code>, <code>np.log</code> : racine carrée, exponentielle, logarithme népérien</li>
        <li><code>np.cos</code>, <code>np.sin</code> : fonctions trigonométriques</li>
        <li><code>np.pi</code>, <code>np.e</code> : constantes $\pi$ et $\text{e}$</li>
    </ul>
</div>

### Maximum, minimum, somme, moyenne, variance, écart-type

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soit A un tableau numpy.
    <ul>
        <li style="margin:1rem;"><code>A.max()</code>/<code>A.min()</code> : <strong>maximum/minimum</strong> de A</li>
        <li style="margin:1rem;"><code>A.argmax()</code>/<code>A.argmin()</code> : <strong>indice du maximum/minimum</strong> de A <i>(tableau aplati)</i></li>
        <li style="margin:1rem;"><code>A.sum()</code> : <strong>somme</strong> des éléments de A</li>
        <li style="margin:1rem;"><code>A.mean()</code>, <code>A.var()</code>, <code>A.std()</code> : <strong>moyenne, variance, écart-type</strong> des éléments de A</li>
        <li style="margin:1rem;"><code>A.max(axis=0)</code>, <code>A.sum(axis=1)</code>, ... : mêmes calculs <strong>selon une direction</strong> <i>(0 : par colonne, 1 : par ligne)</i></li>
    </ul>
</div>

### Booléens

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soit A un tableau numpy.
    <ul>
        <li style="margin:1rem;"><code>A &gt; x</code> : tableau booléen indiquant <strong>quels éléments vérifient la condition</strong> <i>(idem pour <code>&lt;</code>, <code>&lt;=</code>, <code>&gt;=</code>, <code>==</code>, <code>!=</code>)</i></li>
        <li style="margin:1rem;"><code>~</code>, <code>&amp;</code>, <code>|</code> : <strong>négation, et, ou</strong> entre tableaux booléens</li>
        <li style="margin:1rem;"><code>A[A &gt; x]</code> : <strong>sous-tableau des éléments</strong> vérifiant la condition</li>
        <li style="margin:1rem;"><code>A[A &gt; x] = v</code> : <strong>remplace par v</strong> les éléments vérifiant la condition</li>
        <li style="margin:1rem;"><code>np.where(A &gt; x)</code> : indices vérifiant la condition, <strong>sous forme de tuple</strong></li>
        <li style="margin:1rem;"><code>np.argwhere(A &gt; x)</code> : indices vérifiant la condition, <strong>sous forme de tableau</strong></li>
    </ul>
</div>

### Calcul matriciel

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soient A et B deux matrices numpy.
    <ul>
        <li style="margin:1rem;"><code>np.dot(A, B)</code>/<code>A.dot(B)</code> : <strong>produit matriciel</strong> de A et B <i>(sous condition d'existence)</i></li>
        <li style="margin:1rem;"><code>A.T</code> : <strong>transposée</strong> de A</li>
    </ul>
    Le module <code>numpy.linalg</code> (importé usuellement <code>import numpy.linalg as alg</code>) fournit :
    <ul>
        <li style="margin:1rem;"><code>alg.matrix_power(A, n)</code> : <strong>matrice A puissance n</strong></li>
        <li style="margin:1rem;"><code>alg.det(A)</code> : <strong>déterminant</strong> de A</li>
        <li style="margin:1rem;"><code>alg.inv(A)</code> : <strong>inverse</strong> de A <i>(sous condition d'existence)</i></li>
        <li style="margin:1rem;"><code>alg.solve(A, B)</code> : <strong>solution</strong> de l'équation AX = B <i>(sous condition d'existence)</i></li>
    </ul>
</div>

## SymPy

L'import se fait via `import sympy as sp`.

### Symboles et expressions symboliques

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Pour utiliser le calcul formel, il faut d'abord définir des <strong>variables symboliques</strong>, de deux façons possibles :
    <ul>
        <li><code>x, y = sp.symbols('x y')</code> : <strong>crée</strong> des symboles nommés x et y</li>
        <li><code>from sympy.abc import x, y</code> : <strong>importe directement</strong> les symboles x et y</li>
    </ul>
Une <strong>expression symbolique</strong> se construit à partir de variables symboliques et des opérateurs usuels <i>(exemple : <code>expr = x**2 + 2*x + 1</code>)</i>.    
    
</div>

<div style="
    border:1px solid #b8c7d9;

    padding:0.8rem 1rem;

    background:#f3f7fb;
    color:#2b2b2b;
">
    <ul>
        <li><code>sp.pi</code>, <code>sp.E</code>, <code>sp.I</code> : <strong>constantes</strong> $\pi$, e, i</li>
        <li><code>sp.oo</code>/<code>-sp.oo</code> : <strong>+∞/-∞</strong></li>
        <li><code>sp.sqrt</code>, <code>sp.cos</code>, ... : <strong>fonctions usuelles</strong></li>
        <li><code>expr.evalf()</code>/<code>sp.N(expr, n)</code> : <strong>valeur approchée</strong> d'une expression <i>(les calculs sont exacts par défaut)</i></li>
    </ul>
</div>

        <li><code>sp.factor(expr)</code> : <strong>factorise</strong> expr</li>

### Dérivation, intégration, limites

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
Soit expr une expression symbolique dépendant de x.
    <ul>
        <li><code>sp.diff(expr, x)</code> : <strong>dérivée</strong> de expr par rapport à x</li>
        <li><code>sp.diff(expr, x, n)</code> : <strong>dérivée n-ième</strong> de expr par rapport à x</li>
        <li><code>sp.integrate(expr, x)</code> : <strong>primitive</strong> de expr par rapport à x</li>
        <li><code>sp.integrate(expr, (x, a, b))</code> : <strong>intégrale</strong> de expr entre a et b</li>
        <li><code>sp.limit(expr, x, a)</code> : <strong>limite</strong> de expr quand x tend vers a</li>
    </ul>
</div>

### Équations et équations différentielles

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
    <ul>
        <li><code>sp.solve(sp.Eq(expr1, expr2), x)</code>/<code>sp.solve(expr, x)</code> : <strong>résout</strong> une équation <i>(expr = 0)</i></li>
        <li><code>sp.solveset(expr, x, domain=sp.S.Reals)</code> : <strong>résout</strong> une équation sur un domaine donné</li>
        <li><code>sp.solve((eq1, eq2), (x, y))</code> : <strong>résout un système</strong> d'équations</li>
        <li><code>Y = sp.Function('Y')</code> puis <code>sp.dsolve(equation_diff)</code> : <strong>résout une équation différentielle</strong></li>
    </ul>
</div>

    </ul>
