## Syntaxe élémentaire

- Le langage Python est **sensible à la casse et aux indentations**.
- Les *fins d’instructions* sont marquées par un simple **retour à la ligne**.
- Les **indentations** font partie de la syntaxe. Elles permettent de *définir les blocs* sans avoir à utiliser d’instruction de fin de bloc.
- Pour obtenir des *fonctions plus avancées* en langage Python, il est nécessaire d’**importer des bibliothèques/modules** qui contiennent des instructions supplémentaires.
- Les commentaires se font :
    - sur une ligne : en utilisant le symbole `#` en début de ligne ;
    - sur plusieurs lignes : encadrés par des symboles de multilignes `"""` `"""`.
- Les docstrings (documentations de fonctions) se font :
    - sur une ligne : encadrés par des symboles `"` `"`;
    - sur plusieurs lignes : encadrés par des symboles de multilignes `"""` `"""`.

```python
# Itialisation de la variable a
a = 10

# définition de la fonction f
def f(x):
    """
    Fonction qui à tout réel x,
    associe son cube
    """
    return x**3

# Affichage de f(a)
print(f(a))
```

## Opérations entre nombres

<div class="bloc"><div class="encadre">
Soient a et b deux nombres. Sous condition d’existence :
    <ul>
        <li><code>a + b</code> : <strong>somme</strong> de a et b</li>
        <li><code>a - b</code> : <strong>différence</strong> de a et b</li>
        <li><code>a * b</code> : <strong>produit</strong> de a par b</li>
        <li><code>a ** b</code> : a <strong>puissance</strong>  b</li>
        <li><code>a / b</code> : <strong>quotient</strong> de a par b</li>
        <li><code>a // b</code> : <strong>quotient dans la division euclidienne</strong> de a par b</li>
        <li><code>a % b</code> : <strong>reste dans la division euclidienne</strong> de a par b</li>
    </ul>
</div></div>

**Remarque :** On peut obtenir la racine carrée d'un réel positif $a$ par l'instruction `a**(1/2)`, sa racine cubique par l'instruction `a**(1/3)`, ...

## Bibliothèques/modules importants

### Syntaxe élémentaire

<div class="bloc"><div class="encadre">
Soit `biblio` une bibliothèse/un module Python dont `el` est un objet.
    <ul>
        <li><code>import biblio</code> : importe la bibliothèque en gardant son nom propre.
            <ul>
                <li>Appel de <code>el</code> : <code>biblio.el</code></li>
                <li>Tout élement de <code>biblio</code> peut être appelé.</li>
            </ul>
        </li>
        <li><code>import biblio as b</code> : importe la bibliothèque en lui affectant le nom <code>b</code>. 
            <ul>
                <li>Appel de <code>el</code> : <code>b.el</code></li>
                <li>Tout élement de <code>biblio</code> peut être appelé.</li>
            </ul>
        </li>
        <li><code>from biblio import el</code> : importe l'élément <code>el</code> de la bibliothèque uniquement. 
            <ul>
                <li>Appel de <code>el</code> : <code>el</code></li>
                <li>Aucun autre élement de <code>biblio</code> peut être appelé.</li>
            </ul>
        </li>
        <li><code>from biblio import *</code> : importe tous les éléments de la bibliothèque. 
            <ul>
                <li>Appel de <code>el</code> : <code>el</code></li>
                <li>Tout élement de <code>biblio</code> peut être appelé.</li>
            </ul>
        </li>
    </ul>
</div></div>

## Listes

### Création et ajout d'éléments

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>liste = []</code> : <strong>crée</strong> une liste <strong>vide</strong></li>
        <li><code>liste = [a, b, c, d, e]</code> : <strong>crée</strong> une liste <strong>composée des éléments</strong> a, b, c, d et e</li>
        <li><code>liste[i]</code> : <strong>éléments numéro i</strong> de la liste <i>(le premier élément étant liste[0])</i></li>
        <li><code>liste[i:j]</code> : <strong>sous-liste</strong> formée des éléments d’indices i à j-1</li>
        <li><code>liste.copy()</code> : crée une <strong>copie &laquo;superficielle&raquo;</strong> de la liste</li>      
        <li><code>liste1+liste2</code> : <strong> nouvelle liste</strong> obtenue par <strong>ajout des éléments de liste2</strong> à la suite des éléments de liste1</li>
        <li><code>liste1.extend(liste2)</code> :  <strong>ajout des éléments de liste2</strong> à la suite des éléments de liste1 <strong> dans liste1</strong></li>
        <li><code>liste.append(a)</code> :  <strong>ajout de l'élément a</strong> à la suite des éléments de la liste <strong> dans la liste</strong></li>
        <li><code>liste.insert(i, a)</code> :  <strong>ajout de l'élément a</strong> au niveau de l'indice i <strong> dans la liste</strong></li>

</ul>




    
</div></div>

### Modification/supression d'éléments

<div class="bloc"><div class="encadre">
   <ul>
       <li><code>liste[i] = a</code> : <strong>remplace</strong> l’élément d’indice i par a</li>
       <li><code>del liste[i]</code> : <strong>supprime</strong> l’élément numéro i de la liste</li>
       <li><code>liste.remove(a)</code> : <strong>supprime la première occurrence</strong> de a dans la liste</li>
       <li><code>liste.clear()</code> : <strong>supprime tous</strong> les éléments de la liste</li>
       <li><code>liste.pop()</code> : <strong>supprime et renvoie le dernier élément</strong> de la liste</li>
       <li><code>liste.pop(i)</code> : <strong>supprime et renvoie l’élément d’indice i</strong> de la liste</li>
    </ul>
</div></div>

### Tri

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>liste.reverse()</code> : <strong> inverse l’ordre</strong>  des éléments de la liste </li> 
        <li><code>liste.sort()</code> : <strong>réordonne</strong> les éléments de la liste <strong>par ordre croissant, alphabétique, ...</strong></li>
        <li><code>sorted(liste)</code> : crée une <strong>nouvelle liste ordonnée</strong> sans modifier la liste initiale</li>
    </ul>

    
</div></div>

### Divers

<div class="bloc"><div class="encadre">

### Symboles et expressions symboliques

<div class="bloc"><div class="encadre">
Pour utiliser le calcul formel, il faut d'abord définir des <strong>variables symboliques</strong>, de deux façons possibles :
    <ul>
        <li><code>x, y = sp.symbols('x y')</code> : <strong>crée</strong> des symboles nommés x et y</li>
        <li><code>from sympy.abc import x, y</code> : <strong>importe directement</strong> les symboles x et y</li>
    </ul>
Une <strong>expression symbolique</strong> se construit à partir de variables symboliques et des opérateurs usuels <i>(exemple : <code>expr = x**2 + 2*x + 1</code>)</i>.    
    
</div></div>

### Constantes et fonctions mathématiques

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>sp.pi</code>, <code>sp.E</code>, <code>sp.I</code> : <strong>constantes</strong> $\pi$, e, i</li>
        <li><code>sp.oo</code>/<code>-sp.oo</code> : <strong>+∞/-∞</strong></li>
        <li><code>sp.sqrt</code>, <code>sp.cos</code>, ... : <strong>fonctions usuelles</strong></li>
        <li><code>expr.evalf()</code>/<code>sp.N(expr, n)</code> : <strong>valeur approchée</strong> d'une expression <i>(les calculs sont exacts par défaut)</i></li>
    </ul>
</div></div>

### Fractions

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>sp.S(a)/b</code>/<code>sp.Rational(a, b)</code> : <strong>fraction exacte</strong> a/b</li>
        <li><code>sp.cancel(expr)</code> : <strong>simplifie</strong> une fraction rationnelle</li>
        <li><code>sp.together(expr)</code> : met sous <strong>dénominateur commun</strong></li>
        <li><code>sp.apart(expr)</code> : <strong>décompose en éléments simples</strong></li>
    </ul>
</div></div>

### Substitution et simplification

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>expr.subs(x, valeur)</code> : <strong>substitue</strong> x par une valeur dans expr</li>
        <li><code>sp.simplify(expr)</code> : <strong>simplifie</strong> expr</li>
        <li><code>sp.factor(expr)</code> : <strong>factorise</strong> expr</li>
        <li><code>sp.expand(expr)</code> : <strong>développe</strong> expr</li>
    </ul>
</div></div>

### Dérivation, intégration, limites

<div class="bloc"><div class="encadre">
Soit expr une expression symbolique dépendant de x.
    <ul>
        <li><code>sp.diff(expr, x)</code> : <strong>dérivée</strong> de expr par rapport à x</li>
        <li><code>sp.diff(expr, x, n)</code> : <strong>dérivée n-ième</strong> de expr par rapport à x</li>
        <li><code>sp.integrate(expr, x)</code> : <strong>primitive</strong> de expr par rapport à x</li>
        <li><code>sp.integrate(expr, (x, a, b))</code> : <strong>intégrale</strong> de expr entre a et b</li>
        <li><code>sp.limit(expr, x, a)</code> : <strong>limite</strong> de expr quand x tend vers a</li>
    </ul>
</div></div>

### Équations et équations différentielles

<div class="bloc"><div class="encadre">
    <ul>
        <li><code>sp.solve(sp.Eq(expr1, expr2), x)</code>/<code>sp.solve(expr, x)</code> : <strong>résout</strong> une équation <i>(expr = 0)</i></li>
        <li><code>sp.solveset(expr, x, domain=sp.S.Reals)</code> : <strong>résout</strong> une équation sur un domaine donné</li>
        <li><code>sp.solve((eq1, eq2), (x, y))</code> : <strong>résout un système</strong> d'équations</li>
        <li><code>Y = sp.Function('Y')</code> puis <code>sp.dsolve(equation_diff)</code> : <strong>résout une équation différentielle</strong></li>
    </ul>
</div></div>

### Matrices

<div class="bloc"><div class="encadre">
Soient M et N deux matrices sympy, créées via <code>Matrix([[...], [...]])</code>.
