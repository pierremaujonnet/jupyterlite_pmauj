# Synthèse instructions élémentaires en Python

## Syntaxe élémentaire

- Le langage Python est **sensible à la casse et aux indentations**.
- Les *fins d’instructions* sont marquées par un simple **retour à la ligne**.
- Les **indentations** font partie de la syntaxe. Elles permettent de *définir les blocs* sans avoir à utiliser d’instruction de fin de bloc.
- Pour obtenir des *fonctions plus avancées* en langage Python, il est nécessaire d’**importer des bibliothèques/modules** qui contiennent des instructions supplémentaires.
- Les commentaires se font :
    - sur une ligne : en utilisant le symbole `#` en début de ligne ;
    - sur plusieurs lignes : encadrés par des symboles de multilignes `"""` `"""` (si aucune affectation).
- Les docstrings (documentations de fonctions) se font :
    - sur une ligne : encadrés par des symboles `"` `"`;
    - sur plusieurs lignes : encadrés par des symboles de multilignes `"""` `"""`.

```python
# Itialisation de la variable a
a = 10

# définition de la fonction f
def f(x):
    """
    Fonction qui à tout réel x,
    associe son cube
    """
    return x**3

# Affichage de f(a)
print(f(a))
```

Soient a et b deux nombres. Sous condition d’existence :
    <ul>

        <li><code>a * b</code> : <strong>produit</strong> de a par b</li>

        <li><code>a / b</code> : <strong>quotient</strong> de a par b</li>
        <li><code>a // b</code> : <strong>quotient dans la division euclidienne</strong> de a par b</li>
        <li><code>a % b</code> : <strong>reste dans la division euclidienne</strong> de a par b</li>
    </ul></div>

## Listes

### Création et ajout d'éléments

<div style="
    border:1px solid #b8c7d9;
    border-left:4px solid #4a78a8;
    border-radius:0.5rem;
    padding:0.8rem 1rem;
    margin:1rem 0;
    background:#f3f7fb;
    color:#2b2b2b;
">
    <ul>
        <li><code>liste = []</code> : <strong>crée</strong> une liste <strong>vide</strong></li>
        <li><code>liste = [a, b, c, d, e]</code> : <strong>crée</strong> une liste <strong>composée des éléments</strong> a, b, c, d et e</li>
        <li><code>liste[i]</code> : <strong>éléments numéro i</strong> de la liste <i>(le premier élément étant liste[0])</i></li>
        <li><code>liste[i:j]</code> : <strong>sous-liste</strong> formée des éléments d’indices i à j-1</li>
        <li><code>liste.copy()</code> : crée une <strong>copie &laquo;superficielle&raquo;</strong> de la liste</li>      
        <li><code>liste1+liste2</code> : <strong> nouvelle liste</strong> obtenue par <strong>ajout des éléments de liste2</strong> à la suite des éléments de liste1</li>
        <li><code>liste1.extend(liste2)</code> :  <strong>ajout des éléments de liste2</strong> à la suite des éléments de liste1 <strong> dans liste1</strong></li>
        <li><code>liste.append(a)</code> :  <strong>ajout de l'élément a</strong> à la suite des éléments de la liste <strong> dans la liste</strong></li>
        <li><code>liste.insert(i, a)</code> :  <strong>ajout de l'élément a</strong> au niveau de l'indice i <strong> dans la liste</strong></li>
</ul>   
</div>
